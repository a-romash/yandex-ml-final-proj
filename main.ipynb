# Мульти скейл

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from PIL import Image
from sklearn.metrics import f1_score, roc_auc_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import random
import warnings
import gc
from datetime import datetime
warnings.filterwarnings('ignore')

# ============= Configuration =============
class Config:
    train_images = '/content/dataset/train_images'
    train_solution = '/content/dataset/train_solution.csv'
    test_images = '/content/dataset/test_images'

    checkpoint_dir = '/content/drive/MyDrive/deepfake_checkpoints'

    img_size = 256
    batch_size = 64
    epochs = 100
    lr = 1e-3
    weight_decay = 1e-4

    seed = 42

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

os.makedirs(Config.checkpoint_dir, exist_ok=True)
print(f"📁 Checkpoints will be saved to: {Config.checkpoint_dir}")

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = True

set_seed(Config.seed)

# ============= 1. Architecture =============
class ConvBlock(nn.Module):
    def __init__(self, in_ch, out_ch, kernel_size=3, stride=1, padding=1):
        super().__init__()
        self.conv = nn.Conv2d(in_ch, out_ch, kernel_size, stride, padding, bias=False)
        self.bn = nn.BatchNorm2d(out_ch)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x):
        return self.relu(self.bn(self.conv(x)))

class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.gap(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y

class MultiScaleBlock(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.branch1 = nn.Conv2d(in_ch, out_ch, 1, bias=False)
        self.branch3 = nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False)
        self.branch5 = nn.Conv2d(in_ch, out_ch, 5, padding=2, bias=False)

        self.bn1 = nn.BatchNorm2d(out_ch)
        self.bn3 = nn.BatchNorm2d(out_ch)
        self.bn5 = nn.BatchNorm2d(out_ch)

        self.fusion = nn.Sequential(
            nn.Conv2d(out_ch * 3, out_ch, 1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            SEBlock(out_ch)
        )

    def forward(self, x):
        b1 = self.bn1(self.branch1(x))
        b3 = self.bn3(self.branch3(x))
        b5 = self.bn5(self.branch5(x))
        out = torch.cat([b1, b3, b5], dim=1)
        return self.fusion(out)

class DeepfakeDetector(nn.Module):
    def __init__(self, num_classes=1):
        super().__init__()

        self.stem = nn.Sequential(
            ConvBlock(3, 32),
            ConvBlock(32, 32),
            nn.MaxPool2d(2),
            ConvBlock(32, 64),
            nn.MaxPool2d(2)
        )

        self.enc1 = nn.Sequential(MultiScaleBlock(64, 128), nn.MaxPool2d(2))
        self.enc2 = nn.Sequential(MultiScaleBlock(128, 256), nn.MaxPool2d(2))
        self.enc3 = nn.Sequential(MultiScaleBlock(256, 512), nn.MaxPool2d(2))
        self.enc4 = nn.Sequential(MultiScaleBlock(512, 1024), nn.AdaptiveAvgPool2d((2, 2)))

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(1024 * 4, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.1),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.stem(x)
        x = self.enc1(x)
        x = self.enc2(x)
        x = self.enc3(x)
        x = self.enc4(x)
        x = self.classifier(x)
        return x

# ============= 2. Dataset =============
class DeepfakeDataset(Dataset):
    def __init__(self, img_dir, annotations_file=None, transform=None, is_train=True):
        self.img_dir = img_dir
        self.transform = transform
        self.is_train = is_train
        self.images_info = []

        if annotations_file and os.path.exists(annotations_file):
            df = pd.read_csv(annotations_file)

            if 'id' in df.columns and 'target_feature' in df.columns:
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['target_feature']))
            else:
                df = pd.read_csv(annotations_file, header=None, names=['id', 'label'])
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['label']))

            if is_train:
                labels = [l for _, l in self.images_info]
                print(f"Dataset: {labels.count(0)} real, {labels.count(1)} fake, total {len(self.images_info)}")
        else:
            image_files = [f for f in os.listdir(img_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]
            image_ids = sorted([int(os.path.splitext(f)[0]) for f in image_files])
            for img_id in image_ids:
                self.images_info.append((img_id, -1))
            print(f"Test dataset: {len(self.images_info)} images")

    def __len__(self):
        return len(self.images_info)

    def __getitem__(self, idx):
        img_id, label = self.images_info[idx]
        img_name = os.path.join(self.img_dir, f'{img_id}.jpg')

        try:
            image = cv2.imread(img_name)
            if image is None:
                image = np.array(Image.open(img_name))
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        except:
            image = np.zeros((Config.img_size, Config.img_size, 3), dtype=np.uint8)

        if self.transform:
            augmented = self.transform(image=image)
            image = augmented['image']

        if label != -1:
            return image, torch.tensor(label, dtype=torch.float32)
        return image

# ============= 3. Augmentations =============
def get_augmentations(is_train=True):
    if is_train:
        return A.Compose([
            A.Resize(Config.img_size, Config.img_size),
            A.RandomResizedCrop(Config.img_size, Config.img_size, scale=(0.8, 1.0)),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.1),
            A.Rotate(limit=25, p=0.4),
            A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.05, rotate_limit=15, p=0.3),
            A.RandomBrightnessContrast(brightness_limit=0.3, contrast_limit=0.3, p=0.6),
            A.HueSaturationValue(hue_shift_limit=15, sat_shift_limit=30, val_shift_limit=20, p=0.4),
            A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),
            A.GaussianBlur(blur_limit=(3, 7), p=0.2),
            A.ImageCompression(quality_lower=70, quality_upper=100, p=0.2),
            A.CoarseDropout(max_holes=8, max_height=32, max_width=32, fill_value=0, p=0.2),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])
    else:
        return A.Compose([
            A.Resize(Config.img_size, Config.img_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])

# ============= 4. MixUp =============
def mixup_data(x, y, alpha=0.2):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1

    batch_size = x.size()[0]
    index = torch.randperm(batch_size).to(x.device)

    mixed_x = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]

    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

# ============= 5. Training Functions =============
def train_epoch(model, loader, optimizer, criterion, device, scaler, use_mixup=True):
    model.train()
    running_loss = 0
    all_preds, all_labels = [], []
    all_probs = []

    for inputs, labels in tqdm(loader, desc='Training'):
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.unsqueeze(1).to(device, non_blocking=True)

        # Apply MixUp
        if use_mixup and np.random.random() < 0.3:
            inputs, y_a, y_b, lam = mixup_data(inputs, labels)

            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs)
                outputs = outputs.view(-1)
                loss = mixup_criterion(criterion, outputs, y_a.view(-1), y_b.view(-1), lam)
        else:
            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs)
                outputs = outputs.view(-1)
                loss = criterion(outputs, labels.view(-1))

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item()

        # Для метрик используем обычную forward pass без MixUp
        with torch.no_grad():
            outputs_eval = model(inputs)
            probs = torch.sigmoid(outputs_eval.view(-1))
            preds = (probs > 0.5).float()

            all_preds.extend(preds.detach().cpu().numpy())
            all_probs.extend(probs.detach().cpu().numpy())
            all_labels.extend(labels.detach().cpu().numpy())

    train_f1 = f1_score(all_labels, all_preds, zero_division=0)
    train_auc = roc_auc_score(all_labels, all_probs)

    return running_loss / len(loader), train_f1, train_auc

def validate_epoch(model, loader, criterion, device, best_threshold=None):
    model.eval()
    val_loss = 0
    val_preds, val_probs, val_labels = [], [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Validation'):
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.unsqueeze(1).to(device, non_blocking=True)

            with autocast():
                outputs = model(inputs)
                outputs = outputs.view(-1)
                loss = criterion(outputs, labels.view(-1))

            val_loss += loss.item()
            probs = torch.sigmoid(outputs)

            threshold = best_threshold if best_threshold is not None else 0.5
            preds = (probs > threshold).float()

            val_preds.extend(preds.cpu().numpy())
            val_probs.extend(probs.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    val_f1 = f1_score(val_labels, val_preds, zero_division=0)
    val_auc = roc_auc_score(val_labels, val_probs)

    return val_loss / len(loader), val_f1, val_auc

def find_best_threshold(model, loader, device):
    model.eval()
    all_probs, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            outputs = model(inputs)
            probs = torch.sigmoid(outputs.view(-1))
            all_probs.extend(probs.cpu().numpy())
            all_labels.extend(labels.numpy())

    best_f1 = 0
    best_thresh = 0.5

    for thresh in np.linspace(0.1, 0.95, 86):
        preds = (np.array(all_probs) > thresh).astype(int)
        f1 = f1_score(all_labels, preds, zero_division=0)
        if f1 > best_f1:
            best_f1 = f1
            best_thresh = thresh

    return best_thresh, best_f1

# ============= 6. Checkpoint Functions =============
def save_checkpoint(model, optimizer, scheduler, epoch, val_f1, val_auc, threshold, history, is_best=False):
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'val_f1': val_f1,
        'val_auc': val_auc,
        'threshold': threshold,
        'history': history,
        'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    if is_best:
        best_path = os.path.join(Config.checkpoint_dir, 'best_model.pth')
        torch.save(checkpoint, best_path)
        print(f"  💾 Best model saved (F1: {val_f1:.4f})")

    if (epoch + 1) % 5 == 0:
        checkpoint_path = os.path.join(Config.checkpoint_dir, f'checkpoint_epoch_{epoch+1}_f1_{val_f1:.4f}.pth')
        torch.save(checkpoint, checkpoint_path)
        print(f"  💾 Checkpoint saved: epoch_{epoch+1}")

    last_path = os.path.join(Config.checkpoint_dir, 'last_checkpoint.pth')
    torch.save(checkpoint, last_path)

def load_checkpoint(model, optimizer, scheduler, checkpoint_path):
    checkpoint = torch.load(checkpoint_path, map_location='cuda', weights_only=False)
    model.load_state_dict(checkpoint['model_state_dict'])
    optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
    scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
    start_epoch = checkpoint['epoch'] + 1
    best_f1 = checkpoint['val_f1']
    best_auc = checkpoint['val_auc']
    threshold = checkpoint['threshold']
    history = checkpoint.get('history', {'train_loss': [], 'train_f1': [], 'train_auc': [],
                                          'val_loss': [], 'val_f1': [], 'val_auc': [],
                                          'best_threshold': []})

    print(f"✅ Loaded checkpoint from epoch {checkpoint['epoch']+1}, F1: {best_f1:.4f}")
    return start_epoch, best_f1, best_auc, threshold, history

def list_checkpoints():
    if os.path.exists(Config.checkpoint_dir):
        checkpoints = [f for f in os.listdir(Config.checkpoint_dir) if f.endswith('.pth')]
        if checkpoints:
            print("\n📂 Available checkpoints:")
            for ckpt in sorted(checkpoints):
                size = os.path.getsize(os.path.join(Config.checkpoint_dir, ckpt)) / 1e6
                print(f"  - {ckpt} ({size:.1f} MB)")

# ============= 7. Main Training =============
def train(resume_from=None):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"🚀 Device: {device}")

    if torch.cuda.is_available():
        print(f"📊 GPU: {torch.cuda.get_device_name(0)}")

    # Dataset
    print("\n📂 Loading dataset...")
    full_dataset = DeepfakeDataset(
        Config.train_images,
        Config.train_solution,
        get_augmentations(is_train=True),
        is_train=True
    )

    # Split
    indices = list(range(len(full_dataset)))
    labels = [full_dataset.images_info[i][1] for i in indices]
    train_idx, val_idx = train_test_split(indices, test_size=0.1, stratify=labels, random_state=Config.seed)

    train_dataset = torch.utils.data.Subset(full_dataset, train_idx)
    val_dataset = torch.utils.data.Subset(full_dataset, val_idx)

    # Class weights
    train_labels = [full_dataset.images_info[i][1] for i in train_idx]
    class_counts = np.bincount(train_labels)
    pos_weight = torch.tensor([class_counts[0] / class_counts[1]]).to(device)

    print(f"\n📊 Class distribution:")
    print(f"  Train - Real: {class_counts[0]}, Fake: {class_counts[1]}")
    print(f"  Pos weight: {pos_weight.item():.3f}")

    # Weighted sampler
    class_weights = 1.0 / class_counts
    sample_weights = class_weights[train_labels]
    sampler = WeightedRandomSampler(sample_weights, len(sample_weights))

    # Loaders
    train_loader = DataLoader(train_dataset, batch_size=Config.batch_size, sampler=sampler,
                              num_workers=2, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=Config.batch_size, shuffle=False,
                            num_workers=2, pin_memory=True)

    print(f"\n📈 Train: {len(train_dataset)}, Val: {len(val_dataset)}")

    # Model
    model = DeepfakeDetector().to(device)
    print(f"📊 Parameters: {sum(p.numel() for p in model.parameters()):,}")

    # Optimizer
    optimizer = optim.AdamW(model.parameters(), lr=Config.lr, weight_decay=Config.weight_decay)

    # Scheduler
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2, eta_min=1e-6)

    # Loss
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

    # Mixed precision
    scaler = GradScaler()

    # Resume
    start_epoch = 0
    best_f1 = 0
    best_auc = 0
    best_threshold = 0.5
    history = {'train_loss': [], 'train_f1': [], 'train_auc': [],
               'val_loss': [], 'val_f1': [], 'val_auc': [], 'best_threshold': []}

    if resume_from and os.path.exists(resume_from):
        start_epoch, best_f1, best_auc, best_threshold, history = load_checkpoint(
            model, optimizer, scheduler, resume_from
        )

    # Training loop
    for epoch in range(start_epoch, Config.epochs):
        print(f"\n{'='*60}")
        print(f"Epoch {epoch+1}/{Config.epochs}")
        print(f"{'='*60}")

        train_loss, train_f1, train_auc = train_epoch(
            model, train_loader, optimizer, criterion, device, scaler, use_mixup=True
        )

        epoch_threshold, _ = find_best_threshold(model, val_loader, device)
        val_loss, val_f1, val_auc = validate_epoch(
            model, val_loader, criterion, device, epoch_threshold
        )

        scheduler.step()

        history['train_loss'].append(train_loss)
        history['train_f1'].append(train_f1)
        history['train_auc'].append(train_auc)
        history['val_loss'].append(val_loss)
        history['val_f1'].append(val_f1)
        history['val_auc'].append(val_auc)
        history['best_threshold'].append(epoch_threshold)

        print(f"\n📊 Train - Loss: {train_loss:.4f}, F1: {train_f1:.4f}, AUC: {train_auc:.4f}")
        print(f"📊 Val   - Loss: {val_loss:.4f}, F1: {val_f1:.4f}, AUC: {val_auc:.4f}")
        print(f"📊 Threshold: {epoch_threshold:.3f}")
        print(f"📉 LR: {optimizer.param_groups[0]['lr']:.6f}")

        is_best = val_f1 > best_f1
        if is_best:
            best_f1 = val_f1
            best_auc = val_auc
            best_threshold = epoch_threshold

        save_checkpoint(model, optimizer, scheduler, epoch, val_f1, val_auc,
                       epoch_threshold, history, is_best=is_best)

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        if (epoch + 1) % 10 == 0:
            plot_curves(history)

        if val_f1 > 0.995:
            print(f"\n🎯 Target reached! Stopping early...")
            break

    print(f"\n{'='*60}")
    print(f"🎉 TRAINING COMPLETED!")
    print(f"Best F1: {best_f1:.4f}, Best AUC: {best_auc:.4f}")
    print(f"Best threshold: {best_threshold:.3f}")

    return model, best_f1, best_threshold

def plot_curves(history):
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    axes[0, 0].plot(history['train_loss'], label='Train Loss')
    axes[0, 0].plot(history['val_loss'], label='Val Loss')
    axes[0, 0].set_xlabel('Epoch')
    axes[0, 0].set_ylabel('Loss')
    axes[0, 0].legend()
    axes[0, 0].grid(True)

    axes[0, 1].plot(history['train_f1'], label='Train F1')
    axes[0, 1].plot(history['val_f1'], label='Val F1')
    axes[0, 1].set_xlabel('Epoch')
    axes[0, 1].set_ylabel('F1 Score')
    axes[0, 1].legend()
    axes[0, 1].grid(True)

    axes[1, 0].plot(history['train_auc'], label='Train AUC')
    axes[1, 0].plot(history['val_auc'], label='Val AUC')
    axes[1, 0].set_xlabel('Epoch')
    axes[1, 0].set_ylabel('AUC')
    axes[1, 0].legend()
    axes[1, 0].grid(True)

    axes[1, 1].plot(history['best_threshold'], label='Threshold', color='purple')
    axes[1, 1].set_xlabel('Epoch')
    axes[1, 1].set_ylabel('Threshold')
    axes[1, 1].legend()
    axes[1, 1].grid(True)

    plt.tight_layout()
    plt.savefig(os.path.join(Config.checkpoint_dir, 'training_curves.png'), dpi=150)
    plt.show()

def create_submission(model_path=None, threshold=0.5):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"🚀 Creating submission...")

    model = DeepfakeDetector().to(device)

    if model_path is None:
        model_path = os.path.join(Config.checkpoint_dir, 'best_model.pth')

    if os.path.exists(model_path):
        checkpoint = torch.load(model_path, map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        if 'threshold' in checkpoint:
            threshold = checkpoint['threshold']
        print(f"✅ Loaded model with F1: {checkpoint.get('val_f1', 'unknown')}")
    else:
        print(f"❌ Model not found!")
        return None

    model.eval()

    test_dataset = DeepfakeDataset(
        Config.test_images,
        annotations_file=None,
        transform=get_augmentations(is_train=False),
        is_train=False
    )

    test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False, num_workers=2)

    predictions = []
    img_ids = []

    with torch.no_grad():
        for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
            images = images.to(device)
            outputs = model(images)
            probs = torch.sigmoid(outputs.view(-1))
            preds = (probs > threshold).int()
            predictions.extend(preds.cpu().numpy())

            start_idx = idx * 64
            for i in range(len(images)):
                if start_idx + i < len(test_dataset.images_info):
                    img_ids.append(test_dataset.images_info[start_idx + i][0])

    results = pd.DataFrame({'id': img_ids, 'target_feature': predictions})
    results = results.sort_values('id').reset_index(drop=True)
    results.to_csv('submission.csv', index=False)

    fake_count = sum(predictions)
    real_count = len(predictions) - fake_count
    print(f"\n✅ Submission saved!")
    print(f"📊 Real: {real_count}, Fake: {fake_count}")

    return results

# ============= Main =============
if __name__ == '__main__':
    print("="*60)
    print("🚀 DEEPFAKE DETECTOR - FIXED")
    print("="*60)

    list_checkpoints()

    MODE = 'submit'  # 'train', 'resume', 'submit'

    if MODE == 'train':
        model, best_f1, best_threshold = train(resume_from=None)
        submission = create_submission(threshold=best_threshold)
    elif MODE == 'resume':
        last_checkpoint = os.path.join(Config.checkpoint_dir, 'last_checkpoint.pth')
        if os.path.exists(last_checkpoint):
            model, best_f1, best_threshold = train(resume_from=last_checkpoint)
            submission = create_submission(threshold=best_threshold)
        else:
            print("No checkpoint found")
    elif MODE == 'submit':
        submission = create_submission()
```

# Применение focal loss + ансамбль

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
from torchvision.ops import sigmoid_focal_loss
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from PIL import Image
from sklearn.metrics import f1_score, roc_auc_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import random
import warnings
import gc
from datetime import datetime
import json
import traceback

warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

class Config:
    """Конфигурация эксперимента"""
    # Пути к данным
    train_images = '/content/dataset/train_images'
    train_solution = '/content/dataset/train_solution.csv'
    test_images = '/content/dataset/test_images'

    # Пути для сохранения
    checkpoint_dir = '/content/drive/MyDrive/deepfake_checkpoints_ensemble'
    ensemble_dir = '/content/drive/MyDrive/deepfake_ensemble'

    # Параметры обучения
    img_size = 256
    batch_size = 64
    epochs = 100
    lr = 1e-3
    weight_decay = 1e-4

    # Ансамбль
    ensemble_seeds = [42, 123, 456]

    # Focal Loss
    focal_alpha = 0.75
    focal_gamma = 2.0

    # MixUp
    mixup_alpha = 0.2
    mixup_prob = 0.3

    # Другое
    num_workers = 2
    gradient_clip = 1.0


# Монтируем Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# Создаём директории
os.makedirs(Config.checkpoint_dir, exist_ok=True)
os.makedirs(Config.ensemble_dir, exist_ok=True)

print(f"📁 Checkpoints: {Config.checkpoint_dir}")
print(f"📁 Ensemble: {Config.ensemble_dir}")

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def set_seed(seed: int = 42) -> None:
    """Фиксирует seed для воспроизводимости"""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


def get_device() -> torch.device:
    """Возвращает доступное устройство (CUDA или CPU)"""
    return torch.device('cuda' if torch.cuda.is_available() else 'cpu')


def print_model_info(model: nn.Module) -> None:
    """Выводит информацию о модели"""
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"📊 Parameters: {total_params:,} (trainable: {trainable_params:,})")


# ============================================================================
# MODEL ARCHITECTURE
# ============================================================================

class ConvBlock(nn.Module):
    """Свёрточный блок с BatchNorm и ReLU"""
    def __init__(self, in_channels: int, out_channels: int,
                 kernel_size: int = 3, stride: int = 1, padding: int = 1):
        super().__init__()
        self.conv = nn.Conv2d(in_channels, out_channels, kernel_size,
                              stride, padding, bias=False)
        self.bn = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.relu(self.bn(self.conv(x)))


class SEBlock(nn.Module):
    """Squeeze-and-Excitation блок для канального внимания"""
    def __init__(self, channels: int, reduction: int = 16):
        super().__init__()
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        b, c, _, _ = x.shape
        y = self.gap(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y


class MultiScaleBlock(nn.Module):
    """Мультимасштабный блок с Skip Connection"""
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()

        # Три параллельных ветки
        self.conv1 = nn.Conv2d(in_channels, out_channels, 1, bias=False)
        self.conv3 = nn.Conv2d(in_channels, out_channels, 3, padding=1, bias=False)
        self.conv5 = nn.Conv2d(in_channels, out_channels, 5, padding=2, bias=False)

        self.bn1 = nn.BatchNorm2d(out_channels)
        self.bn3 = nn.BatchNorm2d(out_channels)
        self.bn5 = nn.BatchNorm2d(out_channels)

        # Смешивание признаков
        self.fusion = nn.Sequential(
            nn.Conv2d(out_channels * 3, out_channels, 1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            SEBlock(out_channels)
        )

        # Skip connection
        self.skip = (nn.Identity() if in_channels == out_channels
                    else nn.Conv2d(in_channels, out_channels, 1, bias=False))
        self.skip_bn = nn.Identity() if in_channels == out_channels else nn.BatchNorm2d(out_channels)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        identity = self.skip(x)
        if not isinstance(self.skip_bn, nn.Identity):
            identity = self.skip_bn(identity)

        # Мультимасштабные признаки
        out1 = self.bn1(self.conv1(x))
        out3 = self.bn3(self.conv3(x))
        out5 = self.bn5(self.conv5(x))

        out = torch.cat([out1, out3, out5], dim=1)
        out = self.fusion(out)

        return F.relu(out + identity)


class DeepfakeDetector(nn.Module):
    """Основная модель детектора дипфейков"""
    def __init__(self, num_classes: int = 1):
        super().__init__()

        # Stem
        self.stem = nn.Sequential(
            ConvBlock(3, 32),
            ConvBlock(32, 32),
            nn.MaxPool2d(2),
            ConvBlock(32, 64),
            nn.MaxPool2d(2)
        )

        # Энкодер
        self.enc1 = MultiScaleBlock(64, 128)
        self.pool1 = nn.MaxPool2d(2)

        self.enc2 = MultiScaleBlock(128, 256)
        self.pool2 = nn.MaxPool2d(2)

        self.enc3 = MultiScaleBlock(256, 512)
        self.pool3 = nn.MaxPool2d(2)

        self.enc4 = MultiScaleBlock(512, 1024)
        self.pool4 = nn.AdaptiveAvgPool2d((2, 2))

        # Классификатор
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(1024 * 4, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.1),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Linear(128, num_classes)
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = self.stem(x)
        x = self.enc1(x)
        x = self.pool1(x)
        x = self.enc2(x)
        x = self.pool2(x)
        x = self.enc3(x)
        x = self.pool3(x)
        x = self.enc4(x)
        x = self.pool4(x)
        return self.classifier(x)


# ============================================================================
# DATASET
# ============================================================================

class DeepfakeDataset(Dataset):
    """Датасет для детекции дипфейков"""
    def __init__(self, img_dir: str, annotations_file: str = None,
                 transform=None, is_train: bool = True):
        self.img_dir = img_dir
        self.transform = transform
        self.is_train = is_train
        self.images_info = []

        if annotations_file and os.path.exists(annotations_file):
            df = pd.read_csv(annotations_file)

            if 'id' in df.columns and 'target_feature' in df.columns:
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['target_feature']))
            else:
                df = pd.read_csv(annotations_file, header=None, names=['id', 'label'])
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['label']))

            if is_train:
                labels = [l for _, l in self.images_info]
                print(f"Dataset: {labels.count(0)} real, {labels.count(1)} fake, total {len(self.images_info)}")
        else:
            image_files = [f for f in os.listdir(img_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]
            image_ids = sorted([int(os.path.splitext(f)[0]) for f in image_files])
            for img_id in image_ids:
                self.images_info.append((img_id, -1))
            print(f"Test dataset: {len(self.images_info)} images")

    def __len__(self) -> int:
        return len(self.images_info)

    def __getitem__(self, idx: int):
        img_id, label = self.images_info[idx]
        img_name = os.path.join(self.img_dir, f'{img_id}.jpg')

        try:
            image = cv2.imread(img_name)
            if image is None:
                image = np.array(Image.open(img_name))
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        except:
            image = np.zeros((Config.img_size, Config.img_size, 3), dtype=np.uint8)

        if self.transform:
            augmented = self.transform(image=image)
            image = augmented['image']

        if label != -1:
            return image, torch.tensor(label, dtype=torch.float32)
        return image


# ============================================================================
# AUGMENTATIONS
# ============================================================================

def get_augmentations(is_train: bool = True):
    """Возвращает аугментации для обучения или валидации"""
    if is_train:
        return A.Compose([
            A.Resize(Config.img_size, Config.img_size),
            A.RandomResizedCrop(Config.img_size, Config.img_size, scale=(0.8, 1.0)),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.1),
            A.Rotate(limit=25, p=0.4),
            A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.05, rotate_limit=15, p=0.3),
            A.RandomBrightnessContrast(brightness_limit=0.3, contrast_limit=0.3, p=0.6),
            A.HueSaturationValue(hue_shift_limit=15, sat_shift_limit=30, val_shift_limit=20, p=0.4),
            A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),
            A.GaussianBlur(blur_limit=(3, 7), p=0.2),
            A.ImageCompression(quality_lower=70, quality_upper=100, p=0.2),
            A.CoarseDropout(max_holes=8, max_height=32, max_width=32, fill_value=0, p=0.2),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])
    else:
        return A.Compose([
            A.Resize(Config.img_size, Config.img_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])


# ============================================================================
# MIXUP
# ============================================================================

def mixup_data(x: torch.Tensor, y: torch.Tensor, alpha: float = 0.2):
    """Применяет MixUp аугментацию"""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1

    batch_size = x.size(0)
    index = torch.randperm(batch_size).to(x.device)

    mixed_x = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]

    return mixed_x, y_a, y_b, lam


# ============================================================================
# CHECKPOINT FUNCTIONS
# ============================================================================

def save_checkpoint(model: nn.Module, optimizer: optim.Optimizer, scheduler,
                    epoch: int, val_f1: float, val_auc: float, threshold: float,
                    history: dict, seed: int, is_best: bool = False) -> None:
    """Сохраняет чекпоинт модели"""
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'val_f1': val_f1,
        'val_auc': val_auc,
        'threshold': threshold,
        'history': history,
        'seed': seed,
        'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    # Сохраняем лучшую модель
    if is_best:
        best_path = os.path.join(Config.ensemble_dir, f'model_seed_{seed}_best.pth')
        torch.save(checkpoint, best_path)
        print(f"  💾 Best model saved (F1: {val_f1:.4f})")

    # Сохраняем чекпоинт каждые 5 эпох
    if (epoch + 1) % 5 == 0:
        checkpoint_path = os.path.join(Config.ensemble_dir,
                                       f'checkpoint_seed_{seed}_epoch_{epoch+1}_f1_{val_f1:.4f}.pth')
        torch.save(checkpoint, checkpoint_path)
        print(f"  💾 Checkpoint saved: epoch_{epoch+1}")

    # Всегда сохраняем последний чекпоинт
    last_path = os.path.join(Config.ensemble_dir, f'last_checkpoint_seed_{seed}.pth')
    torch.save(checkpoint, last_path)


def load_checkpoint_safe(model: nn.Module, optimizer: optim.Optimizer,
                          scheduler, checkpoint_path: str):
    """Безопасная загрузка чекпоинта с обработкой ошибок"""
    default_return = (0, 0.0, 0.0, 0.5,
                      {'train_loss': [], 'train_f1': [], 'train_auc': [],
                       'val_loss': [], 'val_f1': [], 'val_auc': []}, 42)

    if not os.path.exists(checkpoint_path):
        print(f"⚠️ Checkpoint file not found: {checkpoint_path}")
        print("   Starting training from scratch.")
        return default_return

    try:
        # Проверяем размер файла
        file_size = os.path.getsize(checkpoint_path) / 1e6
        print(f"   File size: {file_size:.1f} MB")

        if file_size < 1:  # Файл меньше 1MB - подозрительно
            print(f"⚠️ Checkpoint file is too small ({file_size:.1f} MB). Might be corrupted.")
            print("   Starting training from scratch.")
            return default_return

        # Пробуем загрузить
        checkpoint = torch.load(checkpoint_path, map_location='cuda', weights_only=False)

        # Проверяем наличие ключей
        required_keys = ['model_state_dict', 'optimizer_state_dict', 'epoch', 'val_f1']
        for key in required_keys:
            if key not in checkpoint:
                print(f"⚠️ Checkpoint missing required key: {key}")
                return default_return

        model.load_state_dict(checkpoint['model_state_dict'])
        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])

        if 'scheduler_state_dict' in checkpoint:
            scheduler.load_state_dict(checkpoint['scheduler_state_dict'])

        start_epoch = checkpoint['epoch'] + 1
        best_f1 = checkpoint.get('val_f1', 0.0)
        best_auc = checkpoint.get('val_auc', 0.0)
        threshold = checkpoint.get('threshold', 0.5)
        history = checkpoint.get('history', default_return[4])
        seed = checkpoint.get('seed', 42)

        print(f"✅ Loaded checkpoint from epoch {checkpoint['epoch']+1}")
        print(f"   Seed: {seed}, Val F1: {best_f1:.4f}, Val AUC: {best_auc:.4f}")

        return start_epoch, best_f1, best_auc, threshold, history, seed

    except Exception as e:
        print(f"❌ Error loading checkpoint: {e}")
        traceback.print_exc()
        print("   Starting training from scratch.")
        return default_return


def list_checkpoints(seed: int = None) -> list:
    """Показывает все доступные чекпоинты"""
    if seed is not None:
        pattern = f'checkpoint_seed_{seed}_epoch_'
        checkpoints = [f for f in os.listdir(Config.ensemble_dir)
                      if f.startswith(pattern) and f.endswith('.pth')]
    else:
        checkpoints = [f for f in os.listdir(Config.ensemble_dir) if f.endswith('.pth')]

    if checkpoints:
        print(f"\n📂 Available checkpoints:")
        for ckpt in sorted(checkpoints):
            size = os.path.getsize(os.path.join(Config.ensemble_dir, ckpt)) / 1e6
            print(f"  - {ckpt} ({size:.1f} MB)")
    else:
        print("No checkpoints found")

    return checkpoints


# ============================================================================
# TRAINING FUNCTIONS
# ============================================================================

def train_epoch(model: nn.Module, loader: DataLoader, optimizer: optim.Optimizer,
                device: torch.device, scaler: GradScaler, use_mixup: bool = True) -> tuple:
    """Обучает одну эпоху"""
    model.train()
    running_loss = 0.0
    all_preds, all_labels, all_probs = [], [], []

    for inputs, labels in tqdm(loader, desc='Training'):
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.unsqueeze(1).to(device, non_blocking=True)

        # MixUp
        if use_mixup and np.random.random() < Config.mixup_prob:
            inputs, y_a, y_b, lam = mixup_data(inputs, labels, Config.mixup_alpha)

            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs).view(-1)
                loss_a = sigmoid_focal_loss(outputs, y_a.view(-1),
                                           alpha=Config.focal_alpha,
                                           gamma=Config.focal_gamma,
                                           reduction='mean')
                loss_b = sigmoid_focal_loss(outputs, y_b.view(-1),
                                           alpha=Config.focal_alpha,
                                           gamma=Config.focal_gamma,
                                           reduction='mean')
                loss = lam * loss_a + (1 - lam) * loss_b
        else:
            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs).view(-1)
                loss = sigmoid_focal_loss(outputs, labels.view(-1),
                                         alpha=Config.focal_alpha,
                                         gamma=Config.focal_gamma,
                                         reduction='mean')

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=Config.gradient_clip)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item()

        # Метрики
        with torch.no_grad():
            probs = torch.sigmoid(outputs)
            preds = (probs > 0.5).float()
            all_preds.extend(preds.detach().cpu().numpy())
            all_probs.extend(probs.detach().cpu().numpy())
            all_labels.extend(labels.detach().cpu().numpy())

    train_f1 = f1_score(all_labels, all_preds, zero_division=0)
    train_auc = roc_auc_score(all_labels, all_probs)

    return running_loss / len(loader), train_f1, train_auc


def validate_epoch(model: nn.Module, loader: DataLoader, device: torch.device,
                   threshold: float = 0.5) -> tuple:
    """Валидация одной эпохи"""
    model.eval()
    val_loss = 0.0
    val_preds, val_probs, val_labels = [], [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Validation'):
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.unsqueeze(1).to(device, non_blocking=True)

            with autocast():
                outputs = model(inputs).view(-1)
                loss = sigmoid_focal_loss(outputs, labels.view(-1),
                                         alpha=Config.focal_alpha,
                                         gamma=Config.focal_gamma,
                                         reduction='mean')

            val_loss += loss.item()
            probs = torch.sigmoid(outputs)
            preds = (probs > threshold).float()

            val_preds.extend(preds.cpu().numpy())
            val_probs.extend(probs.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    val_f1 = f1_score(val_labels, val_preds, zero_division=0)
    val_auc = roc_auc_score(val_labels, val_probs)

    return val_loss / len(loader), val_f1, val_auc


def find_best_threshold(model: nn.Module, loader: DataLoader, device: torch.device) -> tuple:
    """Находит оптимальный порог для максимизации F1"""
    model.eval()
    all_probs, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Finding threshold'):
            inputs = inputs.to(device)
            outputs = model(inputs).view(-1)
            probs = torch.sigmoid(outputs)
            all_probs.extend(probs.cpu().numpy())
            all_labels.extend(labels.numpy())

    best_f1, best_thresh = 0.0, 0.5

    for thresh in np.linspace(0.1, 0.95, 86):
        preds = (np.array(all_probs) > thresh).astype(int)
        f1 = f1_score(all_labels, preds, zero_division=0)
        if f1 > best_f1:
            best_f1, best_thresh = f1, thresh

    return best_thresh, best_f1


# ============================================================================
# SINGLE MODEL TRAINING
# ============================================================================

def train_single_model(seed: int, resume_from: str = None) -> tuple:
    """Обучает одну модель с заданным seed"""
    set_seed(seed)
    device = get_device()

    print(f"\n{'='*60}")
    print(f"🚀 Training model with seed={seed}")
    print(f"💾 Checkpoints every 5 epochs")
    print(f"{'='*60}")

    # Датасет
    full_dataset = DeepfakeDataset(
        Config.train_images,
        Config.train_solution,
        get_augmentations(is_train=True),
        is_train=True
    )

    # Разделение
    indices = list(range(len(full_dataset)))
    labels = [full_dataset.images_info[i][1] for i in indices]
    train_idx, val_idx = train_test_split(indices, test_size=0.1, stratify=labels, random_state=seed)

    train_dataset = torch.utils.data.Subset(full_dataset, train_idx)
    val_dataset = torch.utils.data.Subset(full_dataset, val_idx)

    # Балансировка классов
    train_labels = [full_dataset.images_info[i][1] for i in train_idx]
    class_counts = np.bincount(train_labels)
    sample_weights = 1.0 / class_counts
    sample_weights = sample_weights[train_labels]
    sampler = WeightedRandomSampler(sample_weights, len(sample_weights))

    print(f"\n📊 Class distribution:")
    print(f"  Train - Real: {class_counts[0]}, Fake: {class_counts[1]}")

    # DataLoaders
    train_loader = DataLoader(train_dataset, batch_size=Config.batch_size,
                              sampler=sampler, num_workers=Config.num_workers, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=Config.batch_size,
                            shuffle=False, num_workers=Config.num_workers, pin_memory=True)

    # Модель
    model = DeepfakeDetector().to(device)
    print_model_info(model)

    # Optimizer, Scheduler, Scaler
    optimizer = optim.AdamW(model.parameters(), lr=Config.lr, weight_decay=Config.weight_decay)
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2, eta_min=1e-6)
    scaler = GradScaler()

    # Resume
    start_epoch, best_f1, best_auc, best_threshold, history, _ = load_checkpoint_safe(
        model, optimizer, scheduler, resume_from
    ) if resume_from else (0, 0.0, 0.0, 0.5,
        {'train_loss': [], 'train_f1': [], 'train_auc': [],
         'val_loss': [], 'val_f1': [], 'val_auc': []}, seed)

    # Training loop
    for epoch in range(start_epoch, Config.epochs):
        print(f"\nEpoch {epoch+1}/{Config.epochs}")

        train_loss, train_f1, train_auc = train_epoch(
            model, train_loader, optimizer, device, scaler, use_mixup=True
        )

        # Находим оптимальный порог для валидации
        epoch_threshold, _ = find_best_threshold(model, val_loader, device)
        val_loss, val_f1, val_auc = validate_epoch(model, val_loader, device, epoch_threshold)

        scheduler.step()

        # Сохраняем историю
        for key, val in zip(['train_loss', 'train_f1', 'train_auc',
                             'val_loss', 'val_f1', 'val_auc'],
                            [train_loss, train_f1, train_auc,
                             val_loss, val_f1, val_auc]):
            history[key].append(val)

        print(f"\n📊 Train - Loss: {train_loss:.4f}, F1: {train_f1:.4f}, AUC: {train_auc:.4f}")
        print(f"📊 Val   - Loss: {val_loss:.4f}, F1: {val_f1:.4f}, AUC: {val_auc:.4f}")
        print(f"📊 Threshold: {epoch_threshold:.3f}")
        print(f"📉 LR: {optimizer.param_groups[0]['lr']:.6f}")

        # Сохраняем чекпоинт
        is_best = val_f1 > best_f1
        if is_best:
            best_f1, best_auc, best_threshold = val_f1, val_auc, epoch_threshold

        save_checkpoint(model, optimizer, scheduler, epoch, val_f1, val_auc,
                       epoch_threshold, history, seed, is_best=is_best)

        # Очистка памяти
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        # Визуализация каждые 10 эпох
        if (epoch + 1) % 10 == 0:
            plot_training_curves(history, seed)

        if val_f1 > 0.99:
            print(f"\n🎯 Target reached! Stopping early...")
            break

    print(f"\n✅ Training completed for seed={seed}")
    print(f"Best F1: {best_f1:.4f}, Threshold: {best_threshold:.3f}")

    return model, best_f1, best_threshold


def plot_training_curves(history: dict, seed: int) -> None:
    """Рисует графики обучения"""
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    axes[0, 0].plot(history['train_loss'], label='Train Loss')
    axes[0, 0].plot(history['val_loss'], label='Val Loss')
    axes[0, 0].set_xlabel('Epoch')
    axes[0, 0].set_ylabel('Loss')
    axes[0, 0].legend()
    axes[0, 0].grid(True)

    axes[0, 1].plot(history['train_f1'], label='Train F1')
    axes[0, 1].plot(history['val_f1'], label='Val F1')
    axes[0, 1].set_xlabel('Epoch')
    axes[0, 1].set_ylabel('F1 Score')
    axes[0, 1].legend()
    axes[0, 1].grid(True)

    axes[1, 0].plot(history['train_auc'], label='Train AUC')
    axes[1, 0].plot(history['val_auc'], label='Val AUC')
    axes[1, 0].set_xlabel('Epoch')
    axes[1, 0].set_ylabel('AUC')
    axes[1, 0].legend()
    axes[1, 0].grid(True)

    # График порогов, если есть
    if 'thresholds' in history and history['thresholds']:
        axes[1, 1].plot(history['thresholds'], label='Threshold', color='purple')
        axes[1, 1].set_xlabel('Epoch')
        axes[1, 1].set_ylabel('Threshold')
        axes[1, 1].legend()
        axes[1, 1].grid(True)

    plt.tight_layout()
    plt.savefig(os.path.join(Config.ensemble_dir, f'training_curves_seed_{seed}.png'), dpi=150)
    plt.close()


# ============================================================================
# ENSEMBLE
# ============================================================================

def train_ensemble(resume_seed: int = None, checkpoint_file: str = None) -> list:
    """Обучает ансамбль из 3 моделей"""
    print("\n" + "="*60)
    print("🎯 TRAINING ENSEMBLE")
    print("="*60)

    models_info = []

    for seed in Config.ensemble_seeds:
        resume_path = None

        if resume_seed == seed:
            if checkpoint_file:
                resume_path = os.path.join(Config.ensemble_dir, checkpoint_file)
                if os.path.exists(resume_path):
                    print(f"\n🔄 Resuming seed={seed} from: {checkpoint_file}")
                else:
                    print(f"\n⚠️ Checkpoint {checkpoint_file} not found")
                    resume_path = None

            if not resume_path:
                last_checkpoint = os.path.join(Config.ensemble_dir, f'last_checkpoint_seed_{seed}.pth')
                if os.path.exists(last_checkpoint):
                    resume_path = last_checkpoint
                    print(f"\n🔄 Resuming seed={seed} from last checkpoint")

        model, best_f1, best_threshold = train_single_model(seed, resume_from=resume_path)

        models_info.append({
            'seed': seed,
            'f1': best_f1,
            'threshold': best_threshold,
            'path': os.path.join(Config.ensemble_dir, f'model_seed_{seed}_best.pth')
        })

    # Сохраняем информацию об ансамбле
    ensemble_info = {
        'models': models_info,
        'ensemble_size': len(models_info),
        'seeds': Config.ensemble_seeds,
        'created_at': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    torch.save(ensemble_info, os.path.join(Config.ensemble_dir, 'ensemble_info.pth'))

    print("\n" + "="*60)
    print("🎉 ENSEMBLE TRAINING COMPLETED!")
    print("="*60)
    for info in models_info:
        print(f"Seed {info['seed']}: F1={info['f1']:.4f}, Threshold={info['threshold']:.3f}")

    return models_info


def ensemble_predict(models_info: list, threshold: float = None) -> pd.DataFrame:
    """Ансамблевое предсказание"""
    device = get_device()
    print(f"\n🚀 Making ensemble predictions...")

    models, thresholds = [], []

    for info in models_info:
        model = DeepfakeDetector().to(device)
        checkpoint = torch.load(info['path'], map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        model.eval()
        models.append(model)
        thresholds.append(info['threshold'])
        print(f"✅ Loaded model seed={info['seed']}, F1={info['f1']:.4f}")

    if threshold is None:
        threshold = np.mean(thresholds)
    print(f"📊 Using threshold: {threshold:.3f}")

    # Тестовый датасет
    test_dataset = DeepfakeDataset(
        Config.test_images,
        annotations_file=None,
        transform=get_augmentations(is_train=False),
        is_train=False
    )

    test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False,
                             num_workers=Config.num_workers)

    all_probs, img_ids = [], []

    with torch.no_grad():
        for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
            images = images.to(device)
            avg_probs = torch.zeros(images.size(0)).to(device)

            for model in models:
                outputs = model(images).view(-1)
                avg_probs += torch.sigmoid(outputs)

            avg_probs /= len(models)
            all_probs.extend(avg_probs.cpu().numpy())

            start_idx = idx * 64
            for i in range(len(images)):
                if start_idx + i < len(test_dataset.images_info):
                    img_ids.append(test_dataset.images_info[start_idx + i][0])

    preds = (np.array(all_probs) > threshold).astype(int)
    results = pd.DataFrame({'id': img_ids, 'target_feature': preds})
    results = results.sort_values('id').reset_index(drop=True)
    results.to_csv('ensemble_submission.csv', index=False)

    fake_count = results['target_feature'].sum()
    print(f"\n✅ Submission saved to ensemble_submission.csv")
    print(f"📊 Fake: {fake_count} ({fake_count/len(results):.2%}), Real: {len(results)-fake_count}")

    return results


# ============================================================================
# MAIN
# ============================================================================

if __name__ == '__main__':
    print("="*60)
    print("🚀 DEEPFAKE DETECTOR - ENHANCED VERSION")
    print("="*60)

    # Режимы работы:
    # 'train' - обучить ансамбль с нуля
    # 'resume' - возобновить обучение с чекпоинта
    # 'predict' - только предсказание
    # 'single' - обучить одну модель

    MODE = 'single'  # <-- ИЗМЕНИТЕ ЗДЕСЬ

    if MODE == 'train':
        models_info = train_ensemble()
        submission = ensemble_predict(models_info)

    elif MODE == 'resume':
        print("\n📂 Доступные чекпоинты:")
        list_checkpoints()

        seed_input = input("\nВведите seed для возобновления (42, 123, 456): ")
        try:
            resume_seed = int(seed_input)
        except ValueError:
            print("Неверный ввод, использую seed=42")
            resume_seed = 42

        checkpoint_input = input("Введите имя чекпоинта (или оставьте пустым для последнего): ")
        checkpoint_file = checkpoint_input if checkpoint_input.strip() else None

        models_info = train_ensemble(resume_seed=resume_seed, checkpoint_file=checkpoint_file)
        submission = ensemble_predict(models_info)

    elif MODE == 'predict':
        ensemble_info_path = os.path.join(Config.ensemble_dir, 'ensemble_info.pth')
        if os.path.exists(ensemble_info_path):
            ensemble_info = torch.load(ensemble_info_path, map_location='cpu')
            submission = ensemble_predict(ensemble_info['models'])
        else:
            print("❌ Ансамбль не найден! Сначала обучите модели.")

    elif MODE == 'single':
        seed = int(input("Введите seed для обучения: "))
        model, f1, threshold = train_single_model(seed)

        # Создаём сабмит для одной модели
        test_dataset = DeepfakeDataset(
            Config.test_images,
            annotations_file=None,
            transform=get_augmentations(is_train=False),
            is_train=False
        )
        test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False,
                                 num_workers=Config.num_workers)

        device = get_device()
        model.eval()
        predictions, img_ids = [], []

        with torch.no_grad():
            for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
                images = images.to(device)
                outputs = model(images).view(-1)
                probs = torch.sigmoid(outputs)
                preds = (probs > threshold).int()
                predictions.extend(preds.cpu().numpy())

                start_idx = idx * 64
                for i in range(len(images)):
                    if start_idx + i < len(test_dataset.images_info):
                        img_ids.append(test_dataset.images_info[start_idx + i][0])

        results = pd.DataFrame({'id': img_ids, 'target_feature': predictions})
        results = results.sort_values('id').reset_index(drop=True)
        results.to_csv('single_submission.csv', index=False)
        print(f"\n✅ Submission saved to single_submission.csv")
```

```python
import torch
torch.set_float32_matmul_precision('high')
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from PIL import Image
from sklearn.metrics import f1_score, roc_auc_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2

import os
import random
import warnings
import gc
warnings.filterwarnings('ignore')
```

## Базовые функции и классы

```python
# ============= Set Seeds =============
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

# ============= 6. Robust Dataset =============
class StyleGANDataset(Dataset):
    def __init__(self, img_dir, annotations_file=None, transform=None):
        self.img_dir = img_dir
        self.transform = transform
        self.images_info = []
        self.image_paths = {}

        valid_extensions = {'.jpg', '.jpeg', '.png'}
        for file_name in os.listdir(img_dir):
            file_stem, file_ext = os.path.splitext(file_name)
            if file_ext.lower() not in valid_extensions or not file_stem.isdigit():
                continue
            self.image_paths[int(file_stem)] = os.path.join(img_dir, file_name)

        if annotations_file and os.path.exists(annotations_file):
            df = pd.read_csv(annotations_file)

            if 'id' in df.columns and 'target_feature' in df.columns:
                for _, row in df.iterrows():
                    self.images_info.append((int(row['id']), int(row['target_feature'])))
            elif 'id' in df.columns and 'label' in df.columns:
                for _, row in df.iterrows():
                    self.images_info.append((int(row['id']), int(row['label'])))
            else:
                df = pd.read_csv(annotations_file, header=None, names=['id', 'label'])
                for _, row in df.iterrows():
                    self.images_info.append((int(row['id']), int(row['label'])))

            missing_ids = [img_id for img_id, _ in self.images_info if img_id not in self.image_paths]
            if missing_ids:
                preview = ', '.join(map(str, missing_ids[:5]))
                raise FileNotFoundError(
                    f"Missing {len(missing_ids)} images in {img_dir}. First missing ids: {preview}"
                )

            labels = [label for _, label in self.images_info]
            print(f"Dataset statistics:")
            print(f"  Class 0 (real): {labels.count(0)} images")
            print(f"  Class 1 (fake): {labels.count(1)} images")
            print(f"  Total: {len(self.images_info)} images")
        else:
            for img_id in sorted(self.image_paths):
                self.images_info.append((img_id, -1))
            print(f"Test dataset: {len(self.images_info)} images")

    def __len__(self):
        return len(self.images_info)

    def __getitem__(self, idx):
        img_id, label = self.images_info[idx]
        img_name = self.image_paths.get(img_id)
        if img_name is None:
            raise FileNotFoundError(f"Image id {img_id} is missing in {self.img_dir}")

        try:
            image = np.array(Image.open(img_name).convert('RGB'))
        except Exception as error:
            raise RuntimeError(f"Error loading image {img_name}: {error}") from error

        if self.transform:
            augmented = self.transform(image=image)
            image = augmented['image']

        if label != -1:
            return image, label
        return image

# ============= 7. Augmentations =============
def get_augmentations(is_train=True, img_size=256):
    if is_train:
        return A.Compose([
            A.Resize(img_size, img_size),
            A.HorizontalFlip(p=0.5),
            A.RandomBrightnessContrast(brightness_limit=0.1, contrast_limit=0.1, p=0.2),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])
    else:
        return A.Compose([
            A.Resize(img_size, img_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])

# ============= 8. Loss Functions =============
class FocalLoss(nn.Module):
    def __init__(self, alpha=0.75, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, inputs, targets):
        bce_loss = F.binary_cross_entropy_with_logits(inputs, targets, reduction='none')
        pt = torch.exp(-bce_loss)
        focal_loss = self.alpha * (1 - pt) ** self.gamma * bce_loss
        return focal_loss.mean()

def compute_metrics_at_threshold(labels, probs, threshold):
    labels = np.asarray(labels).astype(int)
    probs = np.asarray(probs)
    preds = (probs >= threshold).astype(int)

    return {
        'f1': f1_score(labels, preds, zero_division=0),
        'precision': precision_score(labels, preds, zero_division=0),
        'recall': recall_score(labels, preds, zero_division=0),
        'positive_rate': float(preds.mean()),
    }

def find_best_f1_threshold_from_probs(labels, probs, thresholds=None):
    labels = np.asarray(labels).astype(int)
    probs = np.asarray(probs)

    if thresholds is None:
        thresholds = np.linspace(0.05, 0.95, 181)

    best_threshold = 0.5
    best_metrics = compute_metrics_at_threshold(labels, probs, best_threshold)

    for threshold in thresholds:
        metrics = compute_metrics_at_threshold(labels, probs, float(threshold))
        if metrics['f1'] > best_metrics['f1']:
            best_threshold = float(threshold)
            best_metrics = metrics

    return best_threshold, best_metrics

# ============= 9. Training Functions =============
def train_epoch(model, dataloader, optimizer, criterion, device):
    model.train()
    running_loss = 0
    all_preds, all_labels = [], []

    for inputs, labels in tqdm(dataloader, desc='Training'):
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.float().unsqueeze(1).to(device, non_blocking=True)

        optimizer.zero_grad()

        outputs = model(inputs)
        outputs = outputs.view(-1)
        labels = labels.view(-1)
        loss = criterion(outputs, labels)

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        running_loss += loss.item()
        preds = (torch.sigmoid(outputs) > 0.5).float()
        all_preds.extend(preds.detach().cpu().numpy())
        all_labels.extend(labels.detach().cpu().numpy())

    train_f1 = f1_score(all_labels, all_preds, zero_division=0)
    return running_loss / len(dataloader), train_f1

def validate_epoch(model, dataloader, criterion, device):
    model.eval()
    val_loss = 0
    val_probs, val_labels = [], []

    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc='Validation'):
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.float().unsqueeze(1).to(device, non_blocking=True)

            outputs = model(inputs)
            outputs = outputs.view(-1)
            labels = labels.view(-1)
            loss = criterion(outputs, labels)

            val_loss += loss.item()
            probs = torch.sigmoid(outputs)
            val_probs.extend(probs.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    metrics_at_0_5 = compute_metrics_at_threshold(val_labels, val_probs, threshold=0.5)
    best_threshold, best_metrics = find_best_f1_threshold_from_probs(val_labels, val_probs)

    try:
        val_auc = roc_auc_score(val_labels, val_probs)
    except ValueError:
        val_auc = float('nan')

    return (
        val_loss / len(dataloader),
        metrics_at_0_5['f1'],
        val_auc,
        best_metrics['f1'],
        best_threshold,
        best_metrics['precision'],
        best_metrics['recall'],
        best_metrics['positive_rate'],
    )

def find_optimal_threshold(model, val_loader, device):
    model.eval()
    all_probs, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs = inputs.to(device)
            outputs = model(inputs)
            probs = torch.sigmoid(outputs.view(-1))
            all_probs.extend(probs.cpu().numpy())
            all_labels.extend(labels.numpy())

    best_thresh, _ = find_best_f1_threshold_from_probs(all_labels, all_probs)
    return best_thresh

def create_submission(model_path, model, test_images_path, output_path='submission.csv', threshold=0.5):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    checkpoint = torch.load(model_path, map_location=device)

    if 'model_state_dict' in checkpoint:
        model.load_state_dict(checkpoint['model_state_dict'])
        print(f"✅ Loaded model with F1: {checkpoint.get('val_f1', 'unknown')}")

    model.eval()

    test_dataset = StyleGANDataset(
        img_dir=test_images_path,
        annotations_file=None,
        transform=get_augmentations(is_train=False, img_size=256)
    )

    test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False, num_workers=2)

    predictions = []
    img_ids = []

    with torch.no_grad():
        for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
            images = images.to(device)
            outputs = model(images)
            probs = torch.sigmoid(outputs.view(-1))
            preds = (probs > threshold).int()

            predictions.extend(preds.cpu().numpy())

            start_idx = idx * 64
            for i in range(len(images)):
                if start_idx + i < len(test_dataset.images_info):
                    img_ids.append(test_dataset.images_info[start_idx + i][0])

    results = pd.DataFrame({'id': img_ids, 'target_feature': predictions})
    results = results.sort_values('id').reset_index(drop=True)
    results.to_csv(output_path, index=False)

    print(f"✅ Submission saved to {output_path}")
    print(f"Class distribution: {results['target_feature'].value_counts().to_dict()}")

    return results
```

```python
# ============= 10. Manual EfficientNet-B0 =============
import math


def _make_divisible(value, divisor=8, min_value=None):
    if min_value is None:
        min_value = divisor
    new_value = max(min_value, int(value + divisor / 2) // divisor * divisor)
    if new_value < 0.9 * value:
        new_value += divisor
    return new_value


class ConvBNAct(nn.Sequential):
    def __init__(self, in_channels, out_channels, kernel_size, stride=1, groups=1, activation_layer=nn.SiLU):
        padding = (kernel_size - 1) // 2
        layers = [
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size,
                stride=stride,
                padding=padding,
                groups=groups,
                bias=False,
            ),
            nn.BatchNorm2d(out_channels, eps=1e-3, momentum=0.01),
        ]
        if activation_layer is not None:
            layers.append(activation_layer(inplace=True))
        super().__init__(*layers)


class SqueezeExcitation(nn.Module):
    def __init__(self, input_channels, squeeze_channels):
        super().__init__()
        self.avgpool = nn.AdaptiveAvgPool2d(1)
        self.reduce = nn.Conv2d(input_channels, squeeze_channels, kernel_size=1)
        self.act = nn.SiLU(inplace=True)
        self.expand = nn.Conv2d(squeeze_channels, input_channels, kernel_size=1)
        self.gate = nn.Sigmoid()

    def forward(self, x):
        scale = self.avgpool(x)
        scale = self.reduce(scale)
        scale = self.act(scale)
        scale = self.expand(scale)
        return x * self.gate(scale)


class DropConnect(nn.Module):
    def __init__(self, drop_prob=0.0):
        super().__init__()
        self.drop_prob = drop_prob

    def forward(self, x):
        if not self.training or self.drop_prob == 0.0:
            return x

        keep_prob = 1.0 - self.drop_prob
        random_tensor = keep_prob + torch.rand((x.shape[0], 1, 1, 1), dtype=x.dtype, device=x.device)
        binary_mask = torch.floor(random_tensor)
        return x * binary_mask / keep_prob


class MBConvConfig:
    def __init__(self, expand_ratio, kernel_size, stride, input_channels, out_channels, num_layers):
        self.expand_ratio = expand_ratio
        self.kernel_size = kernel_size
        self.stride = stride
        self.input_channels = input_channels
        self.out_channels = out_channels
        self.num_layers = num_layers

    @staticmethod
    def adjust_channels(channels, width_mult):
        return _make_divisible(channels * width_mult, 8)

    @staticmethod
    def adjust_depth(num_layers, depth_mult):
        return int(math.ceil(num_layers * depth_mult))


class MBConv(nn.Module):
    def __init__(self, config, drop_connect_rate=0.0):
        super().__init__()
        self.use_residual = config.stride == 1 and config.input_channels == config.out_channels

        expanded_channels = MBConvConfig.adjust_channels(config.input_channels, config.expand_ratio)
        layers = []

        if expanded_channels != config.input_channels:
            layers.append(
                ConvBNAct(config.input_channels, expanded_channels, kernel_size=1)
            )

        layers.append(
            ConvBNAct(
                expanded_channels,
                expanded_channels,
                kernel_size=config.kernel_size,
                stride=config.stride,
                groups=expanded_channels,
            )
        )

        squeeze_channels = max(1, config.input_channels // 4)
        layers.append(SqueezeExcitation(expanded_channels, squeeze_channels))
        layers.append(
            ConvBNAct(expanded_channels, config.out_channels, kernel_size=1, activation_layer=None)
        )

        self.block = nn.Sequential(*layers)
        self.drop_connect = DropConnect(drop_connect_rate)

    def forward(self, x):
        out = self.block(x)
        if self.use_residual:
            out = self.drop_connect(out)
            out = out + x
        return out


class ManualEfficientNetB0(nn.Module):
    def __init__(self, num_classes=1, width_mult=1.0, depth_mult=1.0, dropout=0.2, drop_connect_rate=0.2):
        super().__init__()
        settings = [
            (1, 3, 1, 32, 16, 1),
            (6, 3, 2, 16, 24, 2),
            (6, 5, 2, 24, 40, 2),
            (6, 3, 2, 40, 80, 3),
            (6, 5, 1, 80, 112, 3),
            (6, 5, 2, 112, 192, 4),
            (6, 3, 1, 192, 320, 1),
        ]

        block_configs = [
            MBConvConfig(
                expand_ratio,
                kernel_size,
                stride,
                MBConvConfig.adjust_channels(in_channels, width_mult),
                MBConvConfig.adjust_channels(out_channels, width_mult),
                MBConvConfig.adjust_depth(num_layers, depth_mult),
            )
            for expand_ratio, kernel_size, stride, in_channels, out_channels, num_layers in settings
        ]

        stem_channels = block_configs[0].input_channels
        layers = [ConvBNAct(3, stem_channels, kernel_size=3, stride=2)]

        total_blocks = sum(config.num_layers for config in block_configs)
        block_id = 0
        for config in block_configs:
            stage = []
            for layer_idx in range(config.num_layers):
                input_channels = config.input_channels if layer_idx == 0 else config.out_channels
                stride = config.stride if layer_idx == 0 else 1
                block_config = MBConvConfig(
                    config.expand_ratio,
                    config.kernel_size,
                    stride,
                    input_channels,
                    config.out_channels,
                    1,
                )
                block_drop_rate = drop_connect_rate * block_id / total_blocks
                stage.append(MBConv(block_config, drop_connect_rate=block_drop_rate))
                block_id += 1
            layers.append(nn.Sequential(*stage))

        head_channels = MBConvConfig.adjust_channels(1280, width_mult)
        layers.append(ConvBNAct(block_configs[-1].out_channels, head_channels, kernel_size=1))

        self.features = nn.Sequential(*layers)
        self.avgpool = nn.AdaptiveAvgPool2d(1)
        self.classifier = nn.Sequential(
            nn.Dropout(p=dropout),
            nn.Linear(head_channels, num_classes),
        )

        self._initialize_weights()

    def _initialize_weights(self):
        for module in self.modules():
            if isinstance(module, nn.Conv2d):
                nn.init.kaiming_normal_(module.weight, mode='fan_out')
                if module.bias is not None:
                    nn.init.zeros_(module.bias)
            elif isinstance(module, nn.BatchNorm2d):
                nn.init.ones_(module.weight)
                nn.init.zeros_(module.bias)
            elif isinstance(module, nn.Linear):
                init_range = 1.0 / math.sqrt(module.in_features)
                nn.init.uniform_(module.weight, -init_range, init_range)
                nn.init.zeros_(module.bias)

    def forward_features(self, x):
        return self.features(x)

    def forward(self, x):
        x = self.forward_features(x)
        x = self.avgpool(x)
        x = torch.flatten(x, 1)
        return self.classifier(x)
```

```python
# ============= 11. Manual EfficientNet-B0 Pipeline =============
import copy

BEST_MODEL_NAME = 'best_manual_effb0.pth'
CHECKPOINT_EVERY = 3


def _resolve_manual_checkpoint_path(checkpoint_dir, checkpoint_name_or_path):
    if checkpoint_name_or_path is None:
        return None

    normalized_checkpoint_dir = os.path.normpath(checkpoint_dir)
    normalized_candidate = os.path.normpath(checkpoint_name_or_path)

    if os.path.isabs(checkpoint_name_or_path):
        return checkpoint_name_or_path

    if normalized_candidate == normalized_checkpoint_dir:
        return checkpoint_name_or_path

    if normalized_candidate.startswith(normalized_checkpoint_dir + os.sep):
        return checkpoint_name_or_path

    return os.path.join(checkpoint_dir, checkpoint_name_or_path)


def list_manual_checkpoints(checkpoint_dir='manual_effb0_checkpoints'):
    if not os.path.exists(checkpoint_dir):
        print(f"No checkpoint directory found at {checkpoint_dir}")
        return []

    checkpoints = sorted(
        [filename for filename in os.listdir(checkpoint_dir) if filename.endswith('.pth')]
    )

    if not checkpoints:
        print(f"No checkpoints found in {checkpoint_dir}")
        return []

    print(f"Manual EfficientNet checkpoints in {checkpoint_dir}:")
    for checkpoint_name in checkpoints:
        checkpoint_path = os.path.join(checkpoint_dir, checkpoint_name)
        size_mb = os.path.getsize(checkpoint_path) / 1e6
        print(f"  - {checkpoint_name} ({size_mb:.1f} MB)")

    return checkpoints


def train_manual_efficientnet(
    train_images_path,
    train_solution_path,
    checkpoint_dir='manual_effb0_checkpoints',
    resume_from_checkpoint=None,
):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using device: {device}")

    if not os.path.exists(train_images_path):
        print(f"Error: {train_images_path} does not exist")
        return None, 0.0, 0.5

    if not os.path.exists(train_solution_path):
        print(f"Error: {train_solution_path} does not exist")
        return None, 0.0, 0.5

    os.makedirs(checkpoint_dir, exist_ok=True)

    IMG_SIZE = 256
    BATCH_SIZE = 32
    NUM_EPOCHS = 30
    LEARNING_RATE = 1e-3

    base_dataset = StyleGANDataset(
        img_dir=train_images_path,
        annotations_file=train_solution_path,
        transform=get_augmentations(is_train=False, img_size=IMG_SIZE),
    )

    indices = list(range(len(base_dataset)))
    labels = [base_dataset.images_info[i][1] for i in indices]

    train_idx, val_idx = train_test_split(
        indices,
        test_size=0.1,
        stratify=labels,
        random_state=42,
    )

    train_dataset = copy.copy(base_dataset)
    train_dataset.transform = get_augmentations(is_train=True, img_size=IMG_SIZE)
    val_dataset = copy.copy(base_dataset)
    val_dataset.transform = get_augmentations(is_train=False, img_size=IMG_SIZE)

    train_subset = torch.utils.data.Subset(train_dataset, train_idx)
    val_subset = torch.utils.data.Subset(val_dataset, val_idx)

    train_labels = [base_dataset.images_info[i][1] for i in train_idx]
    class_counts = np.bincount(train_labels)
    pos_weight = torch.tensor([class_counts[0] / class_counts[1]], dtype=torch.float32, device=device)

    print(f"Train samples: {len(train_subset)}")
    print(f"Validation samples: {len(val_subset)}")
    print(f"Training class distribution: {class_counts.tolist()}")
    print(f"Positive class weight: {pos_weight.item():.4f}")

    train_loader = DataLoader(
        train_subset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=2,
        pin_memory=True,
    )
    val_loader = DataLoader(
        val_subset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=2,
        pin_memory=True,
    )

    model = ManualEfficientNetB0(num_classes=1).to(device)
    total_params = sum(parameter.numel() for parameter in model.parameters())
    print(f"Manual EfficientNet-B0 parameters: {total_params:,}")

    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=NUM_EPOCHS,
        eta_min=1e-6,
    )
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

    best_model_path = os.path.join(checkpoint_dir, BEST_MODEL_NAME)
    start_epoch = 0
    best_val_f1 = 0.0
    best_val_auc = 0.0
    best_threshold = 0.5
    train_losses, train_f1s = [], []
    val_losses, val_f1s, val_aucs = [], [], []

    if resume_from_checkpoint:
        checkpoint_path = _resolve_manual_checkpoint_path(checkpoint_dir, resume_from_checkpoint)
        if not os.path.exists(checkpoint_path):
            print(f"Checkpoint {checkpoint_path} not found")
            return None, 0.0, 0.5

        checkpoint = torch.load(checkpoint_path, map_location=device, weights_only=False)
        model.load_state_dict(checkpoint['model_state_dict'])
        # optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        start_epoch = checkpoint.get('epoch', -1) + 1
        best_val_f1 = checkpoint.get('best_val_f1', checkpoint.get('val_f1', 0.0))
        best_val_auc = checkpoint.get('best_val_auc', checkpoint.get('val_auc', 0.0))
        best_threshold = checkpoint.get('best_threshold', 0.5)
        train_losses = checkpoint.get('train_losses', [])
        train_f1s = checkpoint.get('train_f1s', [])
        val_losses = checkpoint.get('val_losses', [])
        val_f1s = checkpoint.get('val_f1s', [])
        val_aucs = checkpoint.get('val_aucs', [])

        if 'scheduler_state_dict' in checkpoint:
            scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
        else:
            for _ in range(start_epoch):
                scheduler.step()

        print(f"Resuming from {checkpoint_path}")
        print(f"Starting epoch: {start_epoch}")
        print(f"Best validation F1 so far: {best_val_f1:.4f}")
        print(f"Best threshold so far: {best_threshold:.3f}")

    for epoch in range(start_epoch, NUM_EPOCHS):
        print(f"\n{'=' * 70}")
        print(f"Epoch {epoch + 1}/{NUM_EPOCHS}")
        print(f"{'=' * 70}")

        train_loss, train_f1 = train_epoch(model, train_loader, optimizer, criterion, device)
        (
            val_loss,
            val_f1_at_0_5,
            val_auc,
            val_best_f1,
            val_best_threshold,
            val_precision,
            val_recall,
            val_positive_rate,
        ) = validate_epoch(model, val_loader, criterion, device)
        scheduler.step()

        train_losses.append(train_loss)
        train_f1s.append(train_f1)
        val_losses.append(val_loss)
        val_f1s.append(val_best_f1)
        val_aucs.append(val_auc)

        current_lr = optimizer.param_groups[0]['lr']
        print(f"Train Loss: {train_loss:.4f}, Train F1@0.5: {train_f1:.4f}")
        print(f"Val Loss: {val_loss:.4f}, Val F1@0.5: {val_f1_at_0_5:.4f}, Val AUC: {val_auc:.4f}")
        print(f"Val Best F1: {val_best_f1:.4f} at threshold {val_best_threshold:.3f}")
        print(f"Val Precision: {val_precision:.4f}, Val Recall: {val_recall:.4f}, Positive Rate: {val_positive_rate:.4f}")
        print(f"Learning Rate: {current_lr:.6f}")

        checkpoint_state = {
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'scheduler_state_dict': scheduler.state_dict(),
            'val_f1': val_best_f1,
            'val_f1_at_0_5': val_f1_at_0_5,
            'val_auc': val_auc,
            'val_precision': val_precision,
            'val_recall': val_recall,
            'val_positive_rate': val_positive_rate,
            'best_val_f1': best_val_f1,
            'best_val_auc': best_val_auc,
            'best_threshold': val_best_threshold,
            'pos_weight': pos_weight.item(),
            'img_size': IMG_SIZE,
            'train_losses': train_losses,
            'train_f1s': train_f1s,
            'val_losses': val_losses,
            'val_f1s': val_f1s,
            'val_aucs': val_aucs,
        }

        if val_best_f1 > best_val_f1 or not os.path.exists(best_model_path):
            best_val_f1 = val_best_f1
            best_val_auc = val_auc
            best_threshold = val_best_threshold
            checkpoint_state['best_val_f1'] = best_val_f1
            checkpoint_state['best_val_auc'] = best_val_auc
            checkpoint_state['best_threshold'] = best_threshold
            torch.save(checkpoint_state, best_model_path)
            print(f"Saved new best model to {best_model_path}")

        if (epoch + 1) % CHECKPOINT_EVERY == 0:
            checkpoint_name = f"checkpoint_epoch_{epoch + 1}_f1_{val_best_f1:.4f}.pth"
            checkpoint_path = os.path.join(checkpoint_dir, checkpoint_name)
            checkpoint_state['best_val_f1'] = best_val_f1
            checkpoint_state['best_val_auc'] = best_val_auc
            torch.save(checkpoint_state, checkpoint_path)
            print(f"Saved checkpoint: {checkpoint_name}")

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    if os.path.exists(best_model_path):
        best_checkpoint = torch.load(best_model_path, map_location=device)
        model.load_state_dict(best_checkpoint['model_state_dict'])
        best_val_f1 = best_checkpoint.get('best_val_f1', best_checkpoint.get('val_f1', best_val_f1))
        best_val_auc = best_checkpoint.get('best_val_auc', best_checkpoint.get('val_auc', best_val_auc))
        best_threshold = best_checkpoint.get('best_threshold', best_threshold)
    else:
        best_threshold = find_optimal_threshold(model, val_loader, device)
        best_checkpoint = {
            'epoch': NUM_EPOCHS - 1,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'scheduler_state_dict': scheduler.state_dict(),
            'val_f1': best_val_f1,
            'val_auc': best_val_auc,
            'best_val_f1': best_val_f1,
            'best_val_auc': best_val_auc,
            'best_threshold': best_threshold,
            'img_size': IMG_SIZE,
            'train_losses': train_losses,
            'train_f1s': train_f1s,
            'val_losses': val_losses,
            'val_f1s': val_f1s,
            'val_aucs': val_aucs,
        }

    best_checkpoint['best_threshold'] = best_threshold
    torch.save(best_checkpoint, best_model_path)

    print(f"\n{'=' * 70}")
    print("MANUAL EFFICIENTNET-B0 TRAINING COMPLETED")
    print(f"Best F1: {best_val_f1:.4f}")
    print(f"Best AUC: {best_val_auc:.4f}")
    print(f"Optimal threshold: {best_threshold:.3f}")

    return model, best_val_f1, best_threshold


def create_manual_efficientnet_submission(
    model_path,
    test_images_path,
    output_path='submission.csv',
    threshold=None,
    checkpoint_dir='manual_effb0_checkpoints',
):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    checkpoint_path = _resolve_manual_checkpoint_path(checkpoint_dir, model_path)

    if not os.path.exists(checkpoint_path):
        print(f"Error: checkpoint {checkpoint_path} not found")
        return None

    checkpoint = torch.load(checkpoint_path, map_location=device)
    img_size = checkpoint.get('img_size', 256)
    threshold = checkpoint.get('best_threshold') if threshold is None else threshold
    if threshold is None:
        threshold = 0.5

    model = ManualEfficientNetB0(num_classes=1).to(device)
    if 'model_state_dict' in checkpoint:
        model.load_state_dict(checkpoint['model_state_dict'])
    else:
        model.load_state_dict(checkpoint)

    print(f"Using device: {device}")
    print(f"Loaded checkpoint: {checkpoint_path}")
    print(f"Submission threshold: {threshold:.3f}")

    test_dataset = StyleGANDataset(
        img_dir=test_images_path,
        annotations_file=None,
        transform=get_augmentations(is_train=False, img_size=img_size),
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=64,
        shuffle=False,
        num_workers=2,
        pin_memory=True,
    )

    model.eval()
    predictions = []
    img_ids = []

    with torch.no_grad():
        for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
            images = images.to(device)
            outputs = model(images)
            probs = torch.sigmoid(outputs.view(-1))
            preds = (probs > threshold).int()

            predictions.extend(preds.cpu().numpy())

            start_idx = idx * 64
            for sample_idx in range(len(images)):
                dataset_idx = start_idx + sample_idx
                if dataset_idx < len(test_dataset.images_info):
                    img_ids.append(test_dataset.images_info[dataset_idx][0])

    results = pd.DataFrame({'id': img_ids, 'target_feature': predictions})
    results = results.sort_values('id').reset_index(drop=True)
    results.to_csv(output_path, index=False)

    print(f"Submission saved to {output_path}")
    print(f"Class distribution: {results['target_feature'].value_counts().to_dict()}")

    return results


TRAIN_IMAGES = '/content/dataset/train_images'
TRAIN_SOLUTION = '/content/dataset/train_solution.csv'
TEST_IMAGES = '/content/dataset/test_images'
CHECKPOINT_DIR = 'manual_effb0_checkpoints'
MODEL_PATH = BEST_MODEL_NAME
MODE = 'resume'  # 'train', 'resume', 'submit', 'list'
THRESHOLD = None
RESUME_CHECKPOINT = "best_manual_effb0.pth"

if MODE == 'list':
    list_manual_checkpoints(CHECKPOINT_DIR)
elif MODE == 'train':
    train_manual_efficientnet(
        train_images_path=TRAIN_IMAGES,
        train_solution_path=TRAIN_SOLUTION,
        checkpoint_dir=CHECKPOINT_DIR,
        resume_from_checkpoint=None,
    )
elif MODE == 'resume':
    if RESUME_CHECKPOINT is None:
        print('Set RESUME_CHECKPOINT to a checkpoint filename or absolute path before resuming.')
    else:
        train_manual_efficientnet(
            train_images_path=TRAIN_IMAGES,
            train_solution_path=TRAIN_SOLUTION,
            checkpoint_dir=CHECKPOINT_DIR,
            resume_from_checkpoint=RESUME_CHECKPOINT,
        )
elif MODE == 'submit':
    create_manual_efficientnet_submission(
        model_path=MODEL_PATH,
        test_images_path=TEST_IMAGES,
        output_path='submission.csv',
        threshold=THRESHOLD,
        checkpoint_dir=CHECKPOINT_DIR,
    )
else:
    print(f"Unsupported MODE: {MODE}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
from torchvision.ops import sigmoid_focal_loss
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from PIL import Image
from sklearn.metrics import f1_score, roc_auc_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import random
import warnings
import gc
from datetime import datetime
import json
import traceback

warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

class Config:
    """Конфигурация эксперимента"""
    # Пути к данным
    train_images = '/content/dataset/train_images'
    train_solution = '/content/dataset/train_solution.csv'
    test_images = '/content/dataset/test_images'

    # Пути для сохранения
    checkpoint_dir = '/content/drive/MyDrive/deepfake_checkpoints_ensemble'
    ensemble_dir = '/content/drive/MyDrive/deepfake_ensemble'

    # Параметры обучения
    img_size = 256
    batch_size = 64
    epochs = 100
    lr = 1e-3
    weight_decay = 1e-4

    # Ансамбль
    ensemble_seeds = [42, 123, 456]

    # Focal Loss
    focal_alpha = 0.75
    focal_gamma = 2.0

    # MixUp
    mixup_alpha = 0.2
    mixup_prob = 0.3

    # Другое
    num_workers = 2
    gradient_clip = 1.0


# Монтируем Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# Создаём директории
os.makedirs(Config.checkpoint_dir, exist_ok=True)
os.makedirs(Config.ensemble_dir, exist_ok=True)

print(f"📁 Checkpoints: {Config.checkpoint_dir}")
print(f"📁 Ensemble: {Config.ensemble_dir}")

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def set_seed(seed: int = 42) -> None:
    """Фиксирует seed для воспроизводимости"""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


def get_device() -> torch.device:
    """Возвращает доступное устройство (CUDA или CPU)"""
    return torch.device('cuda' if torch.cuda.is_available() else 'cpu')


def print_model_info(model: nn.Module) -> None:
    """Выводит информацию о модели"""
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"📊 Parameters: {total_params:,} (trainable: {trainable_params:,})")


# ============================================================================
# MODEL ARCHITECTURE
# ============================================================================

class ConvBlock(nn.Module):
    """Свёрточный блок с BatchNorm и ReLU"""
    def __init__(self, in_channels: int, out_channels: int,
                 kernel_size: int = 3, stride: int = 1, padding: int = 1):
        super().__init__()
        self.conv = nn.Conv2d(in_channels, out_channels, kernel_size,
                              stride, padding, bias=False)
        self.bn = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.relu(self.bn(self.conv(x)))


class SEBlock(nn.Module):
    """Squeeze-and-Excitation блок для канального внимания"""
    def __init__(self, channels: int, reduction: int = 16):
        super().__init__()
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        b, c, _, _ = x.shape
        y = self.gap(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y


class MultiScaleBlockWithSkip(nn.Module):
    """
    Multi-Scale Block с Skip Connection
    """
    def __init__(self, in_ch, out_ch):
        super().__init__()

        self.branch1 = nn.Conv2d(in_ch, out_ch, kernel_size=1, padding=0, bias=False)
        self.branch3 = nn.Conv2d(in_ch, out_ch, kernel_size=3, padding=1, bias=False)
        self.branch5 = nn.Conv2d(in_ch, out_ch, kernel_size=5, padding=2, bias=False)

        self.bn1 = nn.BatchNorm2d(out_ch)
        self.bn3 = nn.BatchNorm2d(out_ch)
        self.bn5 = nn.BatchNorm2d(out_ch)

        # Fusion layer
        self.fusion = nn.Sequential(
            nn.Conv2d(out_ch * 3, out_ch, kernel_size=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            SEBlock(out_ch)
        )

        # Skip connection
        self.skip = nn.Identity() if in_ch == out_ch else nn.Conv2d(in_ch, out_ch, kernel_size=1, bias=False)
        self.skip_bn = nn.BatchNorm2d(out_ch) if in_ch != out_ch else nn.Identity()

    def forward(self, x):
        # Skip connection path
        identity = self.skip(x)
        if not isinstance(self.skip_bn, nn.Identity):
            identity = self.skip_bn(identity)

        # Multi-scale paths
        b1 = self.bn1(self.branch1(x))
        b3 = self.bn3(self.branch3(x))
        b5 = self.bn5(self.branch5(x))

        # Concatenate and fuse
        out = torch.cat([b1, b3, b5], dim=1)
        out = self.fusion(out)

        # Add skip connection and activate
        out = out + identity
        return F.relu(out)


class DeepfakeDetector(nn.Module):
    """Основная модель детектора дипфейков"""
    def __init__(self, num_classes: int = 1):
        super().__init__()

        # Stem
        self.stem = nn.Sequential(
            ConvBlock(3, 32),
            ConvBlock(32, 32),
            nn.MaxPool2d(2),
            ConvBlock(32, 64),
            nn.MaxPool2d(2)
        )

        # Энкодер
        self.enc1 = MultiScaleBlock(64, 128)
        self.pool1 = nn.MaxPool2d(2)

        self.enc2 = MultiScaleBlock(128, 256)
        self.pool2 = nn.MaxPool2d(2)

        self.enc3 = MultiScaleBlock(256, 512)
        self.pool3 = nn.MaxPool2d(2)

        self.enc4 = MultiScaleBlock(512, 1024)
        self.pool4 = nn.AdaptiveAvgPool2d((2, 2))

        # Классификатор
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(1024 * 4, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.1),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Linear(128, num_classes)
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = self.stem(x)
        x = self.enc1(x)
        x = self.pool1(x)
        x = self.enc2(x)
        x = self.pool2(x)
        x = self.enc3(x)
        x = self.pool3(x)
        x = self.enc4(x)
        x = self.pool4(x)
        return self.classifier(x)


# ============================================================================
# DATASET
# ============================================================================

class DeepfakeDataset(Dataset):
    """Датасет для детекции дипфейков"""
    def __init__(self, img_dir: str, annotations_file: str = None,
                 transform=None, is_train: bool = True):
        self.img_dir = img_dir
        self.transform = transform
        self.is_train = is_train
        self.images_info = []

        if annotations_file and os.path.exists(annotations_file):
            df = pd.read_csv(annotations_file)

            if 'id' in df.columns and 'target_feature' in df.columns:
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['target_feature']))
            else:
                df = pd.read_csv(annotations_file, header=None, names=['id', 'label'])
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['label']))

            if is_train:
                labels = [l for _, l in self.images_info]
                print(f"Dataset: {labels.count(0)} real, {labels.count(1)} fake, total {len(self.images_info)}")
        else:
            image_files = [f for f in os.listdir(img_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]
            image_ids = sorted([int(os.path.splitext(f)[0]) for f in image_files])
            for img_id in image_ids:
                self.images_info.append((img_id, -1))
            print(f"Test dataset: {len(self.images_info)} images")

    def __len__(self) -> int:
        return len(self.images_info)

    def __getitem__(self, idx: int):
        img_id, label = self.images_info[idx]
        img_name = os.path.join(self.img_dir, f'{img_id}.jpg')

        try:
            image = cv2.imread(img_name)
            if image is None:
                image = np.array(Image.open(img_name))
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        except:
            image = np.zeros((Config.img_size, Config.img_size, 3), dtype=np.uint8)

        if self.transform:
            augmented = self.transform(image=image)
            image = augmented['image']

        if label != -1:
            return image, torch.tensor(label, dtype=torch.float32)
        return image


# ============================================================================
# AUGMENTATIONS
# ============================================================================

def get_augmentations(is_train: bool = True):
    """Возвращает аугментации для обучения или валидации"""
    if is_train:
        return A.Compose([
            A.Resize(Config.img_size, Config.img_size), # Changed to (Config.img_size, Config.img_size) if A.Resize accepts size as a tuple
            A.RandomResizedCrop(Config.img_size, Config.img_size, scale=(0.8, 1.0)), # Changed to (Config.img_size, Config.img_size) if A.RandomResizedCrop accepts size as a tuple
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.1),
            A.Rotate(limit=25, p=0.4),
            A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.05, rotate_limit=15, p=0.3),
            A.RandomBrightnessContrast(brightness_limit=0.3, contrast_limit=0.3, p=0.6),
            A.HueSaturationValue(hue_shift_limit=15, sat_shift_limit=30, val_shift_limit=20, p=0.4),
            A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),
            A.GaussianBlur(blur_limit=(3, 7), p=0.2),
            A.ImageCompression(quality_lower=70, quality_upper=100, p=0.2),
            A.CoarseDropout(max_holes=8, max_height=32, max_width=32, fill_value=0, p=0.2),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])
    else:
        return A.Compose([
            A.Resize(Config.img_size, Config.img_size), # Changed to (Config.img_size, Config.img_size) if A.Resize accepts size as a tuple
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])


# ============================================================================
# MIXUP
# ============================================================================

def mixup_data(x: torch.Tensor, y: torch.Tensor, alpha: float = 0.2):
    """Применяет MixUp аугментацию"""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1

    batch_size = x.size(0)
    index = torch.randperm(batch_size).to(x.device)

    mixed_x = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]

    return mixed_x, y_a, y_b, lam


# ============================================================================
# CHECKPOINT FUNCTIONS
# ============================================================================

def save_checkpoint(model: nn.Module, optimizer: optim.Optimizer, scheduler,
                    epoch: int, val_f1: float, val_auc: float, threshold: float,
                    history: dict, seed: int, is_best: bool = False) -> None:
    """Сохраняет чекпоинт модели"""
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'val_f1': val_f1,
        'val_auc': val_auc,
        'threshold': threshold,
        'history': history,
        'seed': seed,
        'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    # Сохраняем лучшую модель
    if is_best:
        best_path = os.path.join(Config.ensemble_dir, f'model_seed_{seed}_best.pth')
        torch.save(checkpoint, best_path)
        print(f"  💾 Best model saved (F1: {val_f1:.4f})")

    # Сохраняем чекпоинт каждые 5 эпох
    if (epoch + 1) % 5 == 0:
        checkpoint_path = os.path.join(Config.ensemble_dir,
                                       f'checkpoint_seed_{seed}_epoch_{epoch+1}_f1_{val_f1:.4f}.pth')
        torch.save(checkpoint, checkpoint_path)
        print(f"  💾 Checkpoint saved: epoch_{epoch+1}")

    # Всегда сохраняем последний чекпоинт
    last_path = os.path.join(Config.ensemble_dir, f'last_checkpoint_seed_{seed}.pth')
    torch.save(checkpoint, last_path)


def load_checkpoint_safe(model: nn.Module, optimizer: optim.Optimizer,
                          scheduler, checkpoint_path: str):
    """Безопасная загрузка чекпоинта с преобразованием имён слоёв"""
    default_return = (0, 0.0, 0.0, 0.5,
                      {'train_loss': [], 'train_f1': [], 'train_auc': [],
                       'val_loss': [], 'val_f1': [], 'val_auc': []}, 42)

    if not os.path.exists(checkpoint_path):
        print(f"⚠️ Checkpoint file not found: {checkpoint_path}")
        return default_return

    try:
        file_size = os.path.getsize(checkpoint_path) / 1e6
        print(f"   File size: {file_size:.1f} MB")

        if file_size < 1:
            print(f"⚠️ Checkpoint file too small, starting fresh.")
            return default_return

        checkpoint = torch.load(checkpoint_path, map_location='cuda', weights_only=False)

        # Функция преобразования имён: branch -> conv
        def convert_keys(state_dict):
            new_state_dict = {}
            for key, value in state_dict.items():
                # Меняем branch на conv
                if 'branch1' in key:
                    new_key = key.replace('branch1', 'conv1')
                elif 'branch3' in key:
                    new_key = key.replace('branch3', 'conv3')
                elif 'branch5' in key:
                    new_key = key.replace('branch5', 'conv5')
                else:
                    new_key = key
                new_state_dict[new_key] = value
            return new_state_dict

        # Преобразуем имена в чекпоинте
        converted_state_dict = convert_keys(checkpoint['model_state_dict'])

        # Загружаем с преобразованными именами
        missing, unexpected = model.load_state_dict(converted_state_dict, strict=False)

        if missing:
            print(f"   ⚠️ Missing keys: {len(missing)}")
        if unexpected:
            print(f"   ⚠️ Unexpected keys: {len(unexpected)}")

        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])

        if 'scheduler_state_dict' in checkpoint:
            scheduler.load_state_dict(checkpoint['scheduler_state_dict'])

        start_epoch = checkpoint['epoch'] + 1
        best_f1 = checkpoint.get('val_f1', 0.0)
        best_auc = checkpoint.get('val_auc', 0.0)
        threshold = checkpoint.get('threshold', 0.5)
        history = checkpoint.get('history', default_return[4])
        seed = checkpoint.get('seed', 42)

        print(f"✅ Loaded checkpoint from epoch {checkpoint['epoch']+1}")
        print(f"   Seed: {seed}, Val F1: {best_f1:.4f}")

        return start_epoch, best_f1, best_auc, threshold, history, seed

    except Exception as e:
        print(f"❌ Error loading checkpoint: {e}")
        traceback.print_exc()
        return default_return


def list_checkpoints(seed: int = None) -> list:
    """Показывает все доступные чекпоинты"""
    if seed is not None:
        pattern = f'checkpoint_seed_{seed}_epoch_'
        checkpoints = [f for f in os.listdir(Config.ensemble_dir)
                      if f.startswith(pattern) and f.endswith('.pth')]
    else:
        checkpoints = [f for f in os.listdir(Config.ensemble_dir) if f.endswith('.pth')]

    if checkpoints:
        print(f"\n📂 Available checkpoints:")
        for ckpt in sorted(checkpoints):
            size = os.path.getsize(os.path.join(Config.ensemble_dir, ckpt)) / 1e6
            print(f"  - {ckpt} ({size:.1f} MB)")
    else:
        print("No checkpoints found")

    return checkpoints


# ============================================================================
# TRAINING FUNCTIONS
# ============================================================================

def train_epoch(model: nn.Module, loader: DataLoader, optimizer: optim.Optimizer,
                device: torch.device, scaler: GradScaler, use_mixup: bool = True) -> tuple:
    """Обучает одну эпоху"""
    model.train()
    running_loss = 0.0
    all_preds, all_labels, all_probs = [], [], []

    for inputs, labels in tqdm(loader, desc='Training'):
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.unsqueeze(1).to(device, non_blocking=True)

        # MixUp
        if use_mixup and np.random.random() < Config.mixup_prob:
            inputs, y_a, y_b, lam = mixup_data(inputs, labels, Config.mixup_alpha)

            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs).view(-1)
                loss_a = sigmoid_focal_loss(outputs, y_a.view(-1),
                                           alpha=Config.focal_alpha,
                                           gamma=Config.focal_gamma,
                                           reduction='mean')
                loss_b = sigmoid_focal_loss(outputs, y_b.view(-1),
                                           alpha=Config.focal_alpha,
                                           gamma=Config.focal_gamma,
                                           reduction='mean')
                loss = lam * loss_a + (1 - lam) * loss_b
        else:
            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs).view(-1)
                loss = sigmoid_focal_loss(outputs, labels.view(-1),
                                         alpha=Config.focal_alpha,
                                         gamma=Config.focal_gamma,
                                         reduction='mean')

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=Config.gradient_clip)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item()

        # Метрики
        with torch.no_grad():
            probs = torch.sigmoid(outputs)
            preds = (probs > 0.5).float()
            all_preds.extend(preds.detach().cpu().numpy())
            all_probs.extend(probs.detach().cpu().numpy())
            all_labels.extend(labels.detach().cpu().numpy())

    train_f1 = f1_score(all_labels, all_preds, zero_division=0)
    train_auc = roc_auc_score(all_labels, all_probs)

    return running_loss / len(loader), train_f1, train_auc


def validate_epoch(model: nn.Module, loader: DataLoader, device: torch.device,
                   threshold: float = 0.5) -> tuple:
    """Валидация одной эпохи"""
    model.eval()
    val_loss = 0.0
    val_preds, val_probs, val_labels = [], [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Validation'):
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.unsqueeze(1).to(device, non_blocking=True)

            with autocast():
                outputs = model(inputs).view(-1)
                loss = sigmoid_focal_loss(outputs, labels.view(-1),
                                         alpha=Config.focal_alpha,
                                         gamma=Config.focal_gamma,
                                         reduction='mean')

            val_loss += loss.item()
            probs = torch.sigmoid(outputs)
            preds = (probs > threshold).float()

            val_preds.extend(preds.cpu().numpy())
            val_probs.extend(probs.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    val_f1 = f1_score(val_labels, val_preds, zero_division=0)
    val_auc = roc_auc_score(val_labels, val_probs)

    return val_loss / len(loader), val_f1, val_auc


def find_best_threshold(model: nn.Module, loader: DataLoader, device: torch.device) -> tuple:
    """Находит оптимальный порог для максимизации F1"""
    model.eval()
    all_probs, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Finding threshold'):
            inputs = inputs.to(device)
            outputs = model(inputs).view(-1)
            probs = torch.sigmoid(outputs)
            all_probs.extend(probs.cpu().numpy())
            all_labels.extend(labels.numpy())

    best_f1, best_thresh = 0.0, 0.5

    for thresh in np.linspace(0.1, 0.95, 86):
        preds = (np.array(all_probs) > thresh).astype(int)
        f1 = f1_score(all_labels, preds, zero_division=0)
        if f1 > best_f1:
            best_f1, best_thresh = f1, thresh

    return best_thresh, best_f1


# ============================================================================
# SINGLE MODEL TRAINING
# ============================================================================

def train_single_model(seed: int, resume_from: str = None) -> tuple:
    """Обучает одну модель с заданным seed"""
    set_seed(seed)
    device = get_device()

    print(f"\n{'='*60}")
    print(f"🚀 Training model with seed={seed}")
    print(f"💾 Checkpoints every 5 epochs")
    print(f"{'='*60}")

    # Датасет
    full_dataset = DeepfakeDataset(
        Config.train_images,
        Config.train_solution,
        get_augmentations(is_train=True),
        is_train=True
    )

    # Разделение
    indices = list(range(len(full_dataset)))
    labels = [full_dataset.images_info[i][1] for i in indices]
    train_idx, val_idx = train_test_split(indices, test_size=0.1, stratify=labels, random_state=seed)

    train_dataset = torch.utils.data.Subset(full_dataset, train_idx)
    val_dataset = torch.utils.data.Subset(full_dataset, val_idx)

    # Балансировка классов
    train_labels = [full_dataset.images_info[i][1] for i in train_idx]
    class_counts = np.bincount(train_labels)
    sample_weights = 1.0 / class_counts
    sample_weights = sample_weights[train_labels]
    sampler = WeightedRandomSampler(sample_weights, len(sample_weights))

    print(f"\n📊 Class distribution:")
    print(f"  Train - Real: {class_counts[0]}, Fake: {class_counts[1]}")

    # DataLoaders
    train_loader = DataLoader(train_dataset, batch_size=Config.batch_size,
                              sampler=sampler, num_workers=Config.num_workers, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=Config.batch_size,
                            shuffle=False, num_workers=Config.num_workers, pin_memory=True)

    # Модель
    model = DeepfakeDetector().to(device)
    print_model_info(model)

    # Optimizer, Scheduler, Scaler
    optimizer = optim.AdamW(model.parameters(), lr=Config.lr, weight_decay=Config.weight_decay)
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2, eta_min=1e-6)
    scaler = GradScaler()

    # Resume
    start_epoch, best_f1, best_auc, best_threshold, history, _ = load_checkpoint_safe(
        model, optimizer, scheduler, resume_from
    ) if resume_from else (0, 0.0, 0.0, 0.5,
        {'train_loss': [], 'train_f1': [], 'train_auc': [],
         'val_loss': [], 'val_f1': [], 'val_auc': []}, seed)

    # Training loop
    for epoch in range(start_epoch, Config.epochs):
        print(f"\nEpoch {epoch+1}/{Config.epochs}")

        train_loss, train_f1, train_auc = train_epoch(
            model, train_loader, optimizer, device, scaler, use_mixup=True
        )

        # Находим оптимальный порог для валидации
        epoch_threshold, _ = find_best_threshold(model, val_loader, device)
        val_loss, val_f1, val_auc = validate_epoch(model, val_loader, device, epoch_threshold)

        scheduler.step()

        # Сохраняем историю
        for key, val in zip(['train_loss', 'train_f1', 'train_auc',
                             'val_loss', 'val_f1', 'val_auc'],
                            [train_loss, train_f1, train_auc,
                             val_loss, val_f1, val_auc]):
            history[key].append(val)

        print(f"\n📊 Train - Loss: {train_loss:.4f}, F1: {train_f1:.4f}, AUC: {train_auc:.4f}")
        print(f"📊 Val   - Loss: {val_loss:.4f}, F1: {val_f1:.4f}, AUC: {val_auc:.4f}")
        print(f"📊 Threshold: {epoch_threshold:.3f}")
        print(f"📉 LR: {optimizer.param_groups[0]['lr']:.6f}")

        # Сохраняем чекпоинт
        is_best = val_f1 > best_f1
        if is_best:
            best_f1, best_auc, best_threshold = val_f1, val_auc, epoch_threshold

        save_checkpoint(model, optimizer, scheduler, epoch, val_f1, val_auc,
                       epoch_threshold, history, seed, is_best=is_best)

        # Очистка памяти
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        # Визуализация каждые 10 эпох
        if (epoch + 1) % 10 == 0:
            plot_training_curves(history, seed)

        if val_f1 > 0.99:
            print(f"\n🎯 Target reached! Stopping early...")
            break

    print(f"\n✅ Training completed for seed={seed}")
    print(f"Best F1: {best_f1:.4f}, Threshold: {best_threshold:.3f}")

    return model, best_f1, best_threshold


def plot_training_curves(history: dict, seed: int) -> None:
    """Рисует графики обучения"""
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    axes[0, 0].plot(history['train_loss'], label='Train Loss')
    axes[0, 0].plot(history['val_loss'], label='Val Loss')
    axes[0, 0].set_xlabel('Epoch')
    axes[0, 0].set_ylabel('Loss')
    axes[0, 0].legend()
    axes[0, 0].grid(True)

    axes[0, 1].plot(history['train_f1'], label='Train F1')
    axes[0, 1].plot(history['val_f1'], label='Val F1')
    axes[0, 1].set_xlabel('Epoch')
    axes[0, 1].set_ylabel('F1 Score')
    axes[0, 1].legend()
    axes[0, 1].grid(True)

    axes[1, 0].plot(history['train_auc'], label='Train AUC')
    axes[1, 0].plot(history['val_auc'], label='Val AUC')
    axes[1, 0].set_xlabel('Epoch')
    axes[1, 0].set_ylabel('AUC')
    axes[1, 0].legend()
    axes[1, 0].grid(True)

    # График порогов, если есть
    if 'thresholds' in history and history['thresholds']:
        axes[1, 1].plot(history['thresholds'], label='Threshold', color='purple')
        axes[1, 1].set_xlabel('Epoch')
        axes[1, 1].set_ylabel('Threshold')
    axes[1, 1].legend()
    axes[1, 1].grid(True)

    plt.tight_layout()
    plt.savefig(os.path.join(Config.ensemble_dir, f'training_curves_seed_{seed}.png'), dpi=150)
    plt.close()


# ============================================================================
# ENSEMBLE
# ============================================================================

def train_ensemble(resume_seed: int = None, checkpoint_file: str = None) -> list:
    """Обучает ансамбль из 3 моделей"""
    print("\n" + "="*60)
    print("🎯 TRAINING ENSEMBLE")
    print("="*60)

    models_info = []

    for seed in Config.ensemble_seeds:
        resume_path = None

        if resume_seed == seed:
            if checkpoint_file:
                resume_path = os.path.join(Config.ensemble_dir, checkpoint_file)
                if os.path.exists(resume_path):
                    print(f"\n🔄 Resuming seed={seed} from: {checkpoint_file}")
                else:
                    print(f"\n⚠️ Checkpoint {checkpoint_file} not found")
                    resume_path = None

            if not resume_path:
                last_checkpoint = os.path.join(Config.ensemble_dir, f'last_checkpoint_seed_{seed}.pth')
                if os.path.exists(last_checkpoint):
                    resume_path = last_checkpoint
                    print(f"\n🔄 Resuming seed={seed} from last checkpoint")

        model, best_f1, best_threshold = train_single_model(seed, resume_from=resume_path)

        models_info.append({
            'seed': seed,
            'f1': best_f1,
            'threshold': best_threshold,
            'path': os.path.join(Config.ensemble_dir, f'model_seed_{seed}_best.pth')
        })

    # Сохраняем информацию об ансамбле
    ensemble_info = {
        'models': models_info,
        'ensemble_size': len(models_info),
        'seeds': Config.ensemble_seeds,
        'created_at': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    torch.save(ensemble_info, os.path.join(Config.ensemble_dir, 'ensemble_info.pth'))

    print("\n" + "="*60)
    print("🎉 ENSEMBLE TRAINING COMPLETED!")
    print("="*60)
    for info in models_info:
        print(f"Seed {info['seed']}: F1={info['f1']:.4f}, Threshold={info['threshold']:.3f}")

    return models_info


def ensemble_predict(models_info: list, threshold: float = None) -> pd.DataFrame:
    """Ансамблевое предсказание"""
    device = get_device()
    print(f"\n🚀 Making ensemble predictions...")

    models, thresholds = [], []

    for info in models_info:
        model = DeepfakeDetector().to(device)
        checkpoint = torch.load(info['path'], map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        model.eval()
        models.append(model)
        thresholds.append(info['threshold'])
        print(f"✅ Loaded model seed={info['seed']}, F1={info['f1']:.4f}")

    if threshold is None:
        threshold = np.mean(thresholds)
    print(f"📊 Using threshold: {threshold:.3f}")

    # Тестовый датасет
    test_dataset = DeepfakeDataset(
        Config.test_images,
        annotations_file=None,
        transform=get_augmentations(is_train=False),
        is_train=False
    )

    test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False,
                             num_workers=Config.num_workers)

    all_probs, img_ids = [], []

    with torch.no_grad():
        for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
            images = images.to(device)
            avg_probs = torch.zeros(images.size(0)).to(device)

            for model in models:
                outputs = model(images).view(-1)
                avg_probs += torch.sigmoid(outputs)

            avg_probs /= len(models)
            all_probs.extend(avg_probs.cpu().numpy())

            start_idx = idx * 64
            for i in range(len(images)):
                if start_idx + i < len(test_dataset.images_info):
                    img_ids.append(test_dataset.images_info[start_idx + i][0])

    preds = (np.array(all_probs) > threshold).astype(int)
    results = pd.DataFrame({'id': img_ids, 'target_feature': preds})
    results = results.sort_values('id').reset_index(drop=True)
    results.to_csv('ensemble_submission.csv', index=False)

    fake_count = results['target_feature'].sum()
    print(f"\n✅ Submission saved to ensemble_submission.csv")
    print(f"📊 Fake: {fake_count} ({fake_count/len(results):.2%}), Real: {len(results)-fake_count}")

    return results


# ============================================================================
# MAIN
# ============================================================================

if __name__ == '__main__':
    print("="*60)
    print("🚀 DEEPFAKE DETECTOR - ENHANCED VERSION")
    print("="*60)

    # Режимы работы:
    # 'train' - обучить ансамбль с нуля
    # 'resume' - возобновить обучение с чекпоинта
    # 'predict' - только предсказание
    # 'single' - обучить одну модель

    MODE = 'resume'  # <-- ИЗМЕНИТЕ ЗДЕСЬ

    if MODE == 'train':
        models_info = train_ensemble()
        submission = ensemble_predict(models_info)

    elif MODE == 'resume':
        print("\n📂 Доступные чекпоинты:")
        list_checkpoints()

        seed_input = input("\nВведите seed для возобновления (42, 123, 456): ")
        try:
            resume_seed = int(seed_input)
        except ValueError:
            print("Неверный ввод, использую seed=42")
            resume_seed = 42

        checkpoint_input = input("Введите имя чекпоинта (или оставьте пустым для последнего): ")
        checkpoint_file = checkpoint_input if checkpoint_input.strip() else None

        models_info = train_ensemble(resume_seed=resume_seed, checkpoint_file=checkpoint_file)
        submission = ensemble_predict(models_info)

    elif MODE == 'predict':
        ensemble_info_path = os.path.join(Config.ensemble_dir, 'ensemble_info.pth')
        if os.path.exists(ensemble_info_path):
            ensemble_info = torch.load(ensemble_info_path, map_location='cpu')
            submission = ensemble_predict(ensemble_info['models'])
        else:
            print("❌ Ансамбль не найден! Сначала обучите модели.")

    elif MODE == 'single':
        seed = int(input("Введите seed для обучения: "))
        model, f1, threshold = train_single_model(seed)

        # Создаём сабмит для одной модели
        test_dataset = DeepfakeDataset(
            Config.test_images,
            annotations_file=None,
            transform=get_augmentations(is_train=False),
            is_train=False
        )
        test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False,
                                 num_workers=Config.num_workers)

        device = get_device()
        model.eval()
        predictions, img_ids = [], []

        with torch.no_grad():
            for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
                images = images.to(device)
                outputs = model(images).view(-1)
                probs = torch.sigmoid(outputs)
                preds = (probs > threshold).int()
                predictions.extend(preds.cpu().numpy())

                start_idx = idx * 64
                for i in range(len(images)):
                    if start_idx + i < len(test_dataset.images_info):
                        img_ids.append(test_dataset.images_info[start_idx + i][0])

        results = pd.DataFrame({'id': img_ids, 'target_feature': predictions})
        results = results.sort_values('id').reset_index(drop=True)
        results.to_csv('single_submission.csv', index=False)
        print(f"\n✅ Submission saved to single_submission.csv")
```

```python
def fft_transform(x):
    # x: (B, 3, H, W)
    fft = torch.fft.fft2(x)
    fft = torch.abs(fft)
    return fft
```

```python
class FocalLoss(nn.Module):
    def __init__(self, alpha=0.9, gamma=2):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, inputs, targets):
        bce = nn.functional.binary_cross_entropy_with_logits(inputs, targets, reduction='none')
        pt = torch.exp(-bce)
        loss = self.alpha * (1 - pt) ** self.gamma * bce
        return loss.mean()
```

```python
import torch
import matplotlib.pyplot as plt
from sklearn.metrics import f1_score

def train_model(model, train_loader, val_loader, criterion, optimizer, num_epochs, device):
    train_losses, val_losses = [], []
    train_f1_scores, val_f1_scores = [], []

    for epoch in range(num_epochs):
        # ------------------- TRAIN -------------------
        model.train()
        train_loss = 0
        all_preds, all_labels = [], []

        for batch_idx, (inputs, labels) in enumerate(train_loader):
            inputs = inputs.to(device)
            labels = labels.float().unsqueeze(1).to(device)

            optimizer.zero_grad()


            outputs = model(inputs)

            loss = criterion(outputs, labels)

            loss.backward()
            optimizer.step()

            train_loss += loss.item()

            preds = (torch.sigmoid(outputs) > 0.5).float()
            all_preds.append(preds.detach().cpu())
            all_labels.append(labels.detach().cpu())

            if (batch_idx+1) % 100 == 0:
                avg_loss = train_loss / (batch_idx+1)
                print(f"Epoch {epoch+1} | Batch {batch_idx+1} | Avg Loss: {avg_loss:.4f}")

        all_preds_tensor = torch.cat(all_preds)
        all_labels_tensor = torch.cat(all_labels)
        train_f1 = f1_score(all_labels_tensor.numpy(), all_preds_tensor.numpy())

        train_losses.append(train_loss / len(train_loader))
        train_f1_scores.append(train_f1)


        model.eval()
        val_loss = 0
        all_preds, all_labels = [], []

        with torch.no_grad():
            for batch_idx, (inputs, labels) in enumerate(val_loader):
                inputs = inputs.to(device)
                labels = labels.float().unsqueeze(1).to(device)


                outputs = model(inputs)

                loss = criterion(outputs, labels)
                val_loss += loss.item()

                preds = (torch.sigmoid(outputs) > 0.5).float()
                all_preds.append(preds.cpu())
                all_labels.append(labels.cpu())

                if (batch_idx+1) % 100 == 0:
                    avg_loss = val_loss / (batch_idx+1)
                    print(f"Epoch {epoch+1} | Val Batch {batch_idx+1} | Avg Loss: {avg_loss:.4f}")

        all_preds_tensor = torch.cat(all_preds)
        all_labels_tensor = torch.cat(all_labels)
        val_f1 = f1_score(all_labels_tensor.numpy(), all_preds_tensor.numpy())

        val_losses.append(val_loss / len(val_loader))
        val_f1_scores.append(val_f1)

        print(f"Epoch {epoch+1} finished: Train F1={train_f1:.4f} | Val F1={val_f1:.4f}")

    # ------------------- Графики -------------------
    plt.figure(figsize=(10,4))

    plt.subplot(1,2,1)
    plt.plot(train_losses,label="Train Loss")
    plt.plot(val_losses,label="Val Loss")
    plt.legend()
    plt.title("Loss")

    plt.subplot(1,2,2)
    plt.plot(train_f1_scores,label="Train F1")
    plt.plot(val_f1_scores,label="Val F1")
    plt.legend()
    plt.title("F1 Score")

    plt.show()

    return model
```

```python
class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, dropout=0.2):
        super().__init__()

        self.conv1 = nn.Conv2d(in_channels, out_channels, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU()

        self.conv2 = nn.Conv2d(out_channels, out_channels, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(out_channels)

        self.dropout = nn.Dropout2d(dropout)

        # skip connection (если меняется число каналов)
        if in_channels != out_channels:
            self.skip = nn.Conv2d(in_channels, out_channels, 1)
        else:
            self.skip = nn.Identity()

    def forward(self, x):
        identity = self.skip(x)

        out = self.conv1(x)
        out = self.bn1(out)
        out = self.relu(out)

        out = self.dropout(out)

        out = self.conv2(out)
        out = self.bn2(out)

        out += identity
        out = self.relu(out)

        return out

class DualEfficientNetV2(nn.Module):
    def __init__(self):
        super().__init__()

        # 🔹 RGB ветка
        self.rgb_model = efficientnet_b0(weights=None)
        self.rgb_model.classifier = nn.Identity()  # [B, 1280]

        # 🔹 FFT ветка
        self.fft_model = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),

            ResidualBlock(32, 64, dropout=0.2),
            nn.MaxPool2d(2),

            ResidualBlock(64, 128, dropout=0.3),
            nn.MaxPool2d(2),

            ResidualBlock(128, 128, dropout=0.3),

            nn.AdaptiveAvgPool2d(1)
        )

        # 🔹 Г
        self.gate = nn.Sequential(
            nn.Linear(1280, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.Sigmoid()
        )

        # 🔹 Классификатор
        self.fc1 = nn.Linear(1280 + 128, 256)
        self.fc2 = nn.Linear(256, 256)
        self.fc_out = nn.Linear(256, 1)

        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.5)

    def forward(self, x):
        # RGB
        rgb_features = self.rgb_model(x)  # [B, 1280]

        # FFT
        fft = fft_transform(x)
        fft_features = self.fft_model(fft)  # [B, 128, 1, 1]
        fft_features = fft_features.view(fft_features.size(0), -1)  # [B, 128]

        # 🔹 Гейтинг
        gate = self.gate(rgb_features)
        fft_features = fft_features * gate

        # concat
        features = torch.cat([rgb_features, fft_features], dim=1)


        x1 = self.fc1(features)
        x1 = self.relu(x1)
        x1 = self.dropout(x1)

        x2 = self.fc2(x1)
        x2 = self.relu(x2)
        x2 = self.dropout(x2)

        x = x1 + x2  # skip connection

        out = self.fc_out(x)
        return out
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
from torchvision.ops import sigmoid_focal_loss
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from PIL import Image
from sklearn.metrics import f1_score, roc_auc_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import random
import warnings
import gc
from datetime import datetime
import json
import traceback

warnings.filterwarnings('ignore')

# ============================================================================
# КОНФИГУРАЦИЯ
# ============================================================================

class Config:
    # Пути к данным
    train_images = '/content/dataset/train_images'
    train_solution = '/content/dataset/train_solution.csv'
    test_images = '/content/dataset/test_images'

    # Пути для сохранения
    checkpoint_dir = '/content/drive/MyDrive/deepfake_checkpoints_ensemble'
    ensemble_dir = '/content/drive/MyDrive/deepfake_ensemble'

    # Параметры обучения
    img_size = 256
    batch_size = 64
    epochs = 150
    lr = 1e-3
    weight_decay = 5e-4

    # Focal Loss
    focal_alpha = 0.75
    focal_gamma = 2.0

    # MixUp (усиленный)
    mixup_alpha = 0.3
    mixup_prob = 0.5

    # Progressive Resizing
    use_progressive_resizing = True
    img_size_schedule = {
        0: 256,
        31: 288,
        61: 320,
        91: 384,
    }

    # Другое
    num_workers = 2
    gradient_clip = 1.0

    # Ансамбль
    ensemble_seeds = [42, 123, 456]

# Монтируем Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# Создаём директории
os.makedirs(Config.checkpoint_dir, exist_ok=True)
os.makedirs(Config.ensemble_dir, exist_ok=True)

print(f"📁 Checkpoints: {Config.checkpoint_dir}")
print(f"📁 Ensemble: {Config.ensemble_dir}")

# ============================================================================
# ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ============================================================================

def set_seed(seed: int = 42) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


def get_device() -> torch.device:
    return torch.device('cuda' if torch.cuda.is_available() else 'cpu')


def print_model_info(model: nn.Module) -> None:
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"📊 Parameters: {total_params:,} (trainable: {trainable_params:,})")


# ============================================================================
# АРХИТЕКТУРА МОДЕЛИ (ВАША ПРОВЕРЕННАЯ)
# ============================================================================

class ConvBlock(nn.Module):
    def __init__(self, in_channels: int, out_channels: int,
                 kernel_size: int = 3, stride: int = 1, padding: int = 1):
        super().__init__()
        self.conv = nn.Conv2d(in_channels, out_channels, kernel_size,
                              stride, padding, bias=False)
        self.bn = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.relu(self.bn(self.conv(x)))


class SEBlock(nn.Module):
    def __init__(self, channels: int, reduction: int = 16):
        super().__init__()
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        b, c, _, _ = x.shape
        y = self.gap(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y


class MultiScaleBlockWithSkip(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()

        self.branch1 = nn.Conv2d(in_ch, out_ch, kernel_size=1, padding=0, bias=False)
        self.branch3 = nn.Conv2d(in_ch, out_ch, kernel_size=3, padding=1, bias=False)
        self.branch5 = nn.Conv2d(in_ch, out_ch, kernel_size=5, padding=2, bias=False)

        self.bn1 = nn.BatchNorm2d(out_ch)
        self.bn3 = nn.BatchNorm2d(out_ch)
        self.bn5 = nn.BatchNorm2d(out_ch)

        self.fusion = nn.Sequential(
            nn.Conv2d(out_ch * 3, out_ch, kernel_size=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            SEBlock(out_ch)
        )

        self.skip = nn.Identity() if in_ch == out_ch else nn.Conv2d(in_ch, out_ch, kernel_size=1, bias=False)
        self.skip_bn = nn.BatchNorm2d(out_ch) if in_ch != out_ch else nn.Identity()

    def forward(self, x):
        identity = self.skip(x)
        if not isinstance(self.skip_bn, nn.Identity):
            identity = self.skip_bn(identity)

        b1 = self.bn1(self.branch1(x))
        b3 = self.bn3(self.branch3(x))
        b5 = self.bn5(self.branch5(x))

        out = torch.cat([b1, b3, b5], dim=1)
        out = self.fusion(out)
        out = out + identity
        return F.relu(out)


class DeepfakeDetector(nn.Module):
    def __init__(self, num_classes: int = 1):
        super().__init__()

        self.stem = nn.Sequential(
            ConvBlock(3, 32),
            ConvBlock(32, 32),
            nn.MaxPool2d(2),
            ConvBlock(32, 64),
            nn.MaxPool2d(2)
        )

        self.enc1 = MultiScaleBlockWithSkip(64, 128)
        self.pool1 = nn.MaxPool2d(2)

        self.enc2 = MultiScaleBlockWithSkip(128, 256)
        self.pool2 = nn.MaxPool2d(2)

        self.enc3 = MultiScaleBlockWithSkip(256, 512)
        self.pool3 = nn.MaxPool2d(2)

        self.enc4 = MultiScaleBlockWithSkip(512, 1024)
        self.pool4 = nn.AdaptiveAvgPool2d((2, 2))

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(1024 * 4, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.1),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Linear(128, num_classes)
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = self.stem(x)
        x = self.enc1(x)
        x = self.pool1(x)
        x = self.enc2(x)
        x = self.pool2(x)
        x = self.enc3(x)
        x = self.pool3(x)
        x = self.enc4(x)
        x = self.pool4(x)
        return self.classifier(x)


# ============================================================================
# ДАТАСЕТ
# ============================================================================

class DeepfakeDataset(Dataset):
    def __init__(self, img_dir: str, annotations_file: str = None,
                 transform=None, is_train: bool = True):
        self.img_dir = img_dir
        self.transform = transform
        self.is_train = is_train
        self.images_info = []

        if annotations_file and os.path.exists(annotations_file):
            df = pd.read_csv(annotations_file)

            if 'id' in df.columns and 'target_feature' in df.columns:
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['target_feature']))
            else:
                df = pd.read_csv(annotations_file, header=None, names=['id', 'label'])
                for _, row in df.iterrows():
                    self.images_info.append((row['id'], row['label']))

            if is_train:
                labels = [l for _, l in self.images_info]
                print(f"Dataset: {labels.count(0)} real, {labels.count(1)} fake, total {len(self.images_info)}")
        else:
            image_files = [f for f in os.listdir(img_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]
            image_ids = sorted([int(os.path.splitext(f)[0]) for f in image_files])
            for img_id in image_ids:
                self.images_info.append((img_id, -1))
            print(f"Test dataset: {len(self.images_info)} images")

    def __len__(self) -> int:
        return len(self.images_info)

    def __getitem__(self, idx: int):
        img_id, label = self.images_info[idx]
        img_name = os.path.join(self.img_dir, f'{img_id}.jpg')

        try:
            image = cv2.imread(img_name)
            if image is None:
                image = np.array(Image.open(img_name))
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        except:
            image = np.zeros((Config.img_size, Config.img_size, 3), dtype=np.uint8)

        if self.transform:
            augmented = self.transform(image=image)
            image = augmented['image']

        if label != -1:
            return image, torch.tensor(label, dtype=torch.float32)
        return image


# ============================================================================
# АУГМЕНТАЦИИ
# ============================================================================

def get_augmentations(is_train: bool = True, img_size: int = None):
    if img_size is None:
        img_size = Config.img_size

    if is_train:
        return A.Compose([
            A.Resize(height=img_size, width=img_size),
            A.RandomResizedCrop(height=img_size, width=img_size, scale=(0.8, 1.0)),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.1),
            A.Rotate(limit=25, p=0.4),
            A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.05, rotate_limit=15, p=0.3),
            A.RandomBrightnessContrast(brightness_limit=0.3, contrast_limit=0.3, p=0.6),
            A.HueSaturationValue(hue_shift_limit=15, sat_shift_limit=30, val_shift_limit=20, p=0.4),
            A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),
            A.GaussianBlur(blur_limit=(3, 7), p=0.2),
            A.ImageCompression(quality_lower=70, quality_upper=100, p=0.2),
            A.CoarseDropout(max_holes=8, max_height=32, max_width=32, fill_value=0, p=0.2),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])
    else:
        return A.Compose([
            A.Resize(height=img_size, width=img_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])


# ============================================================================
# MIXUP
# ============================================================================

def mixup_data(x: torch.Tensor, y: torch.Tensor, alpha: float = 0.3):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1

    batch_size = x.size(0)
    index = torch.randperm(batch_size).to(x.device)

    mixed_x = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]

    return mixed_x, y_a, y_b, lam


# ============================================================================
# ЧЕКПОИНТЫ
# ============================================================================

def save_checkpoint(model: nn.Module, optimizer: optim.Optimizer, scheduler,
                    epoch: int, val_f1: float, val_auc: float, threshold: float,
                    history: dict, seed: int, is_best: bool = False) -> None:
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'val_f1': val_f1,
        'val_auc': val_auc,
        'threshold': threshold,
        'history': history,
        'seed': seed,
        'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    if is_best:
        best_path = os.path.join(Config.ensemble_dir, f'model_seed_{seed}_best.pth')
        torch.save(checkpoint, best_path)
        print(f"  💾 Best model saved (F1: {val_f1:.4f})")

    if (epoch + 1) % 5 == 0:
        checkpoint_path = os.path.join(Config.ensemble_dir,
                                       f'checkpoint_seed_{seed}_epoch_{epoch+1}_f1_{val_f1:.4f}.pth')
        torch.save(checkpoint, checkpoint_path)
        print(f"  💾 Checkpoint saved: epoch_{epoch+1}")

    last_path = os.path.join(Config.ensemble_dir, f'last_checkpoint_seed_{seed}.pth')
    torch.save(checkpoint, last_path)


def load_checkpoint(model: nn.Module, optimizer: optim.Optimizer,
                    scheduler, checkpoint_path: str):
    """Загружает чекпоинт"""
    if not os.path.exists(checkpoint_path):
        print(f"⚠️ Checkpoint not found: {checkpoint_path}")
        return None, None, None, None, None, None

    # Added weights_only=False to resolve UnpicklingError
    checkpoint = torch.load(checkpoint_path, map_location='cuda', weights_only=False)
    model.load_state_dict(checkpoint['model_state_dict'])
    optimizer.load_state_dict(checkpoint['optimizer_state_dict'])

    if 'scheduler_state_dict' in checkpoint:
        scheduler.load_state_dict(checkpoint['scheduler_state_dict'])

    start_epoch = checkpoint['epoch'] + 1
    best_f1 = checkpoint.get('val_f1', 0.0)
    best_auc = checkpoint.get('val_auc', 0.0)
    threshold = checkpoint.get('threshold', 0.5)
    history = checkpoint.get('history', {
        'train_loss': [], 'train_f1': [], 'train_auc': [],
        'val_loss': [], 'val_f1': [], 'val_auc': []
    })
    seed = checkpoint.get('seed', 42)

    print(f"✅ Loaded checkpoint from epoch {checkpoint['epoch']+1}")
    print(f"   Seed: {seed}, Val F1: {best_f1:.4f}")

    return start_epoch, best_f1, best_auc, threshold, history, seed


def list_checkpoints(seed: int = None) -> list:
    if seed is not None:
        pattern = f'checkpoint_seed_{seed}_epoch_'
        checkpoints = [f for f in os.listdir(Config.ensemble_dir)
                      if f.startswith(pattern) and f.endswith('.pth')]
    else:
        checkpoints = [f for f in os.listdir(Config.ensemble_dir) if f.endswith('.pth')]

    if checkpoints:
        print(f"\n📂 Available checkpoints:")
        for ckpt in sorted(checkpoints):
            size = os.path.getsize(os.path.join(Config.ensemble_dir, ckpt)) / 1e6
            print(f"  - {ckpt} ({size:.1f} MB)")
    else:
        print("No checkpoints found")

    return checkpoints


# ============================================================================
# ФУНКЦИИ ОБУЧЕНИЯ
# ============================================================================

def train_epoch(model: nn.Module, loader: DataLoader, optimizer: optim.Optimizer,
                device: torch.device, scaler: GradScaler) -> tuple:
    model.train()
    running_loss = 0.0
    all_preds, all_labels, all_probs = [], [], []

    for inputs, labels in tqdm(loader, desc='Training'):
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.unsqueeze(1).to(device, non_blocking=True)

        if np.random.random() < Config.mixup_prob:
            inputs, y_a, y_b, lam = mixup_data(inputs, labels, Config.mixup_alpha)

            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs).view(-1)
                loss_a = sigmoid_focal_loss(outputs, y_a.view(-1),
                                           alpha=Config.focal_alpha,
                                           gamma=Config.focal_gamma,
                                           reduction='mean')
                loss_b = sigmoid_focal_loss(outputs, y_b.view(-1),
                                           alpha=Config.focal_alpha,
                                           gamma=Config.focal_gamma,
                                           reduction='mean')
                loss = lam * loss_a + (1 - lam) * loss_b
        else:
            optimizer.zero_grad()
            with autocast():
                outputs = model(inputs).view(-1)
                loss = sigmoid_focal_loss(outputs, labels.view(-1),
                                         alpha=Config.focal_alpha,
                                         gamma=Config.focal_gamma,
                                         reduction='mean')

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=Config.gradient_clip)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item()

        with torch.no_grad():
            probs = torch.sigmoid(outputs)
            preds = (probs > 0.5).float()
            all_preds.extend(preds.detach().cpu().numpy())
            all_probs.extend(probs.detach().cpu().numpy())
            all_labels.extend(labels.detach().cpu().numpy())

    train_f1 = f1_score(all_labels, all_preds, zero_division=0)
    train_auc = roc_auc_score(all_labels, all_probs)

    return running_loss / len(loader), train_f1, train_auc


def validate_epoch(model: nn.Module, loader: DataLoader, device: torch.device,
                   threshold: float = 0.5) -> tuple:
    model.eval()
    val_loss = 0.0
    val_preds, val_probs, val_labels = [], [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Validation'):
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.unsqueeze(1).to(device, non_blocking=True)

            with autocast():
                outputs = model(inputs).view(-1)
                loss = sigmoid_focal_loss(outputs, labels.view(-1),
                                         alpha=Config.focal_alpha,
                                         gamma=Config.focal_gamma,
                                         reduction='mean')

            val_loss += loss.item()
            probs = torch.sigmoid(outputs)
            preds = (probs > threshold).float()

            val_preds.extend(preds.cpu().numpy())
            val_probs.extend(probs.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    val_f1 = f1_score(val_labels, val_preds, zero_division=0)
    val_auc = roc_auc_score(val_labels, val_probs)

    return val_loss / len(loader), val_f1, val_auc


def find_best_threshold(model: nn.Module, loader: DataLoader, device: torch.device) -> tuple:
    model.eval()
    all_probs, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Finding threshold'):
            inputs = inputs.to(device)
            outputs = model(inputs).view(-1)
            probs = torch.sigmoid(outputs)
            all_probs.extend(probs.cpu().numpy())
            all_labels.extend(labels.numpy())

    best_f1, best_thresh = 0.0, 0.5

    for thresh in np.linspace(0.1, 0.95, 86):
        preds = (np.array(all_probs) > thresh).astype(int)
        f1 = f1_score(all_labels, preds, zero_division=0)
        if f1 > best_f1:
            best_f1, best_thresh = f1, thresh

    return best_thresh, best_f1


# ============================================================================
# TTA (TEST TIME AUGMENTATION)
# ============================================================================

class TTA:
    def __init__(self, model: nn.Module, device: torch.device):
        self.model = model
        self.device = device

        self.transforms = [
            lambda x: x,
            lambda x: torch.flip(x, dims=[-1]),
            lambda x: torch.flip(x, dims=[-2]),
            lambda x: torch.rot90(x, k=1, dims=[-2, -1]),
            lambda x: torch.rot90(x, k=2, dims=[-2, -1]),
            lambda x: torch.rot90(x, k=3, dims=[-2, -1]),
            lambda x: torch.rot90(torch.flip(x, dims=[-1]), k=1),
            lambda x: torch.rot90(torch.flip(x, dims=[-2]), k=1),
        ]

    @torch.no_grad()
    def predict(self, images: torch.Tensor) -> torch.Tensor:
        batch_size = images.size(0)
        all_probs = torch.zeros(batch_size).to(self.device)

        for transform in self.transforms:
            augmented = transform(images)
            outputs = self.model(augmented).view(-1)
            all_probs += torch.sigmoid(outputs)

        return all_probs / len(self.transforms)


# ============================================================================
# ОБУЧЕНИЕ ОДНОЙ МОДЕЛИ (С ВОЗМОЖНОСТЬЮ РЕЗЮМЕ)
# ============================================================================

def train_single_model(seed: int, resume_from: str = None,
                       start_epoch: int = 0, best_f1: float = 0.0,
                       best_threshold: float = 0.5, history: dict = None):
    """
    Обучает одну модель с заданным seed
    resume_from: путь к чекпоинту для продолжения обучения
    """
    set_seed(seed)
    device = get_device()

    print(f"\n{'='*60}")
    print(f"🚀 Training model with seed={seed}")
    if resume_from:
        print(f"   Resuming from: {os.path.basename(resume_from)}")
    print(f"💾 Checkpoints every 5 epochs")
    print(f"{'='*60}")

    # Датасет
    full_dataset = DeepfakeDataset(
        Config.train_images,
        Config.train_solution,
        get_augmentations(is_train=True, img_size=Config.img_size),
        is_train=True
    )

    # Разделение
    indices = list(range(len(full_dataset)))
    labels = [full_dataset.images_info[i][1] for i in indices]
    train_idx, val_idx = train_test_split(indices, test_size=0.1, stratify=labels, random_state=seed)

    train_dataset = torch.utils.data.Subset(full_dataset, train_idx)
    val_dataset = torch.utils.data.Subset(full_dataset, val_idx)

    # Балансировка
    train_labels = [full_dataset.images_info[i][1] for i in train_idx]
    class_counts = np.bincount(train_labels)
    sample_weights = 1.0 / class_counts
    sample_weights = sample_weights[train_labels]
    sampler = WeightedRandomSampler(sample_weights, len(sample_weights))

    print(f"\n📊 Class distribution:")
    print(f"  Train - Real: {class_counts[0]}, Fake: {class_counts[1]}")

    # DataLoaders
    train_loader = DataLoader(train_dataset, batch_size=Config.batch_size,
                              sampler=sampler, num_workers=Config.num_workers, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=Config.batch_size,
                            shuffle=False, num_workers=Config.num_workers, pin_memory=True)

    # Модель
    model = DeepfakeDetector().to(device)
    print_model_info(model)

    # Optimizer, Scheduler, Scaler
    optimizer = optim.AdamW(model.parameters(), lr=Config.lr, weight_decay=Config.weight_decay)
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2, eta_min=1e-6)
    scaler = GradScaler()

    # Если передан чекпоинт - загружаем
    if resume_from and os.path.exists(resume_from):
        loaded_start_epoch, loaded_best_f1, loaded_best_auc, loaded_threshold, loaded_history, loaded_seed = load_checkpoint(
            model, optimizer, scheduler, resume_from
        )
        if loaded_start_epoch is not None:
            start_epoch = loaded_start_epoch
            best_f1 = loaded_best_f1
            best_threshold = loaded_threshold
            history = loaded_history
            print(f"   Continuing from epoch {start_epoch}")

    # Если история не передана - создаём новую
    if history is None:
        history = {
            'train_loss': [], 'train_f1': [], 'train_auc': [],
            'val_loss': [], 'val_f1': [], 'val_auc': []
        }

    # Training loop
    for epoch in range(start_epoch, Config.epochs):
        # Progressive Resizing
        if Config.use_progressive_resizing:
            current_img_size = Config.img_size # Store current size for comparison
            for threshold_epoch, new_size in sorted(Config.img_size_schedule.items()):
                if epoch >= threshold_epoch:
                    Config.img_size = new_size

            # Only update transforms if image size has actually changed
            if current_img_size != Config.img_size:
                # Update augmentations for new image size
                full_dataset.transform = get_augmentations(is_train=True, img_size=Config.img_size)
                # Re-create subsets to apply new transforms (or update their transforms if possible directly)
                # For Subset, transform needs to be updated on the original dataset object
                train_dataset.dataset.transform = get_augmentations(is_train=True, img_size=Config.img_size)
                val_dataset.dataset.transform = get_augmentations(is_train=False, img_size=Config.img_size)
                print(f"🖼️ Image size updated to {Config.img_size} for epoch {epoch+1}")

        print(f"\nEpoch {epoch+1}/{Config.epochs}")

        train_loss, train_f1, train_auc = train_epoch(
            model, train_loader, optimizer, device, scaler
        )

        epoch_threshold, _ = find_best_threshold(model, val_loader, device)
        val_loss, val_f1, val_auc = validate_epoch(model, val_loader, device, epoch_threshold)

        scheduler.step()

        # Сохраняем историю
        history['train_loss'].append(train_loss)
        history['train_f1'].append(train_f1)
        history['train_auc'].append(train_auc)
        history['val_loss'].append(val_loss)
        history['val_f1'].append(val_f1)
        history['val_auc'].append(val_auc)

        print(f"\n📊 Train - Loss: {train_loss:.4f}, F1: {train_f1:.4f}, AUC: {train_auc:.4f}")
        print(f"📊 Val   - Loss: {val_loss:.4f}, F1: {val_f1:.4f}, AUC: {val_auc:.4f}")
        print(f"📊 Threshold: {epoch_threshold:.3f}")
        print(f"📉 LR: {optimizer.param_groups[0]['lr']:.6f}")

        is_best = val_f1 > best_f1
        if is_best:
            best_f1 = val_f1
            best_threshold = epoch_threshold

        save_checkpoint(model, optimizer, scheduler, epoch, val_f1, val_auc,
                       epoch_threshold, history, seed, is_best=is_best)

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        if val_f1 > 0.99:
            print(f"\n🎯 Target reached! Stopping early...")
            break

    print(f"\n✅ Training completed for seed={seed}")
    print(f"Best F1: {best_f1:.4f}, Threshold: {best_threshold:.3f}")

    return model, best_f1, best_threshold


# ============================================================================
# ОБУЧЕНИЕ АНСАМБЛЯ
# ============================================================================

def train_ensemble(resume_config: dict = None):
    """
    Обучает ансамбль из 3 моделей

    resume_config: словарь с настройками возобновления, например:
    {
        42: '/path/to/checkpoint_seed_42_epoch_30.pth',
        123: None,  # None = начать с нуля
        456: '/path/to/checkpoint_seed_456_epoch_50.pth'
    }
    """
    print("\n" + "="*60)
    print("🎯 TRAINING ENSEMBLE OF 3 MODELS")
    print("="*60)

    models_info = []

    for seed in Config.ensemble_seeds:
        resume_path = None
        if resume_config and seed in resume_config:
            resume_path = resume_config[seed]
            if resume_path:
                print(f"\n🔄 Will resume seed={seed} from: {os.path.basename(resume_path)}")

        model, best_f1, best_threshold = train_single_model(seed, resume_from=resume_path)

        models_info.append({
            'seed': seed,
            'f1': best_f1,
            'threshold': best_threshold,
            'path': os.path.join(Config.ensemble_dir, f'model_seed_{seed}_best.pth')
        })

    # Сохраняем информацию об ансамбле
    ensemble_info = {
        'models': models_info,
        'ensemble_size': len(models_info),
        'seeds': Config.ensemble_seeds,
        'created_at': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    torch.save(ensemble_info, os.path.join(Config.ensemble_dir, 'ensemble_info.pth'))

    print("\n" + "="*60)
    print("🎉 ENSEMBLE TRAINING COMPLETED!")
    print("="*60)
    for info in models_info:
        print(f"Seed {info['seed']}: F1={info['f1']:.4f}, Threshold={info['threshold']:.3f}")

    return models_info


# ============================================================================
# ПРЕДСКАЗАНИЕ АНСАМБЛЕМ С TTA
# ============================================================================

def ensemble_predict_with_tta(models_info: list, threshold: float = None) -> pd.DataFrame:
    """Ансамблевое предсказание с TTA для каждой модели"""
    device = get_device()
    print(f"\n🚀 Making ensemble predictions with TTA...")

    models_tta = []
    thresholds = []

    for info in models_info:
        model = DeepfakeDetector().to(device)
        checkpoint = torch.load(info['path'], map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        model.eval()

        tta = TTA(model, device)
        models_tta.append(tta)
        thresholds.append(info['threshold'])
        print(f"✅ Loaded model seed={info['seed']}, F1={info['f1']:.4f}")

    if threshold is None:
        threshold = np.mean(thresholds)
    print(f"📊 Using threshold: {threshold:.3f}")

    # Тестовый датасет
    test_dataset = DeepfakeDataset(
        Config.test_images,
        annotations_file=None,
        transform=get_augmentations(is_train=False, img_size=256),
        is_train=False
    )

    test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False,
                             num_workers=Config.num_workers)

    all_probs, img_ids = [], []

    with torch.no_grad():
        for idx, images in enumerate(tqdm(test_loader, desc='Predicting')):
            images = images.to(device)
            avg_probs = torch.zeros(images.size(0)).to(device)

            for tta in models_tta:
                probs = tta.predict(images)
                avg_probs += probs

            avg_probs /= len(models_tta)
            all_probs.extend(avg_probs.cpu().numpy())

            start_idx = idx * 32
            for i in range(len(images)):
                if start_idx + i < len(test_dataset.images_info):
                    img_ids.append(test_dataset.images_info[start_idx + i][0])

    preds = (np.array(all_probs) > threshold).astype(int)
    results = pd.DataFrame({'id': img_ids, 'target_feature': preds})
    results = results.sort_values('id').reset_index(drop=True)
    results.to_csv('ensemble_tta_submission.csv', index=False)

    fake_count = results['target_feature'].sum()
    print(f"\n✅ Submission saved to ensemble_tta_submission.csv")
    print(f"📊 Fake: {fake_count} ({fake_count/len(results):.2%}), Real: {len(results)-fake_count}")

    return results


# ============================================================================
# MAIN
# ============================================================================

if __name__ == '__main__':
    print("="*60)
    print("🚀 DEEPFAKE DETECTOR - FLEXIBLE ENSEMBLE SYSTEM")
    print("="*60)

    # ========================================================================
    # НАСТРОЙКИ - ИЗМЕНИТЕ ЗДЕСЕСЬ
    # ========================================================================

    # Режим работы:
    # 'resume_single' - продолжить обучение одной модели
    # 'train_ensemble' - обучить весь ансамбль
    # 'predict_only' - только предсказание (без обучения)

    MODE = 'resume_single'  # <-- ИЗМЕНИТЕ ЗДЕСЬ

    # ------------------------------------------------------------------------
    # НАСТРОЙКИ ДЛЯ РЕЖИМА 'resume_single'
    # ------------------------------------------------------------------------
    if MODE == 'resume_single':
        # Выберите seed для продолжения
        SEED_TO_RESUME = 42  # 42, 123 или 456

        # Укажите путь к чекпоинту (или оставьте None для автоматического поиска)
        CHECKPOINT_PATH = '/content/drive/MyDrive/deepfake_ensemble/last_checkpoint_seed_42 (5).pth'  # Пример: '/content/drive/MyDrive/deepfake_ensemble/checkpoint_seed_42_epoch_30_f1_0.7743.pth'

        if CHECKPOINT_PATH is None:
            # Автоматически ищем лучший чекпоинт для этого seed
            list_checkpoints(seed=SEED_TO_RESUME)
            print(f"\n⚠️ No checkpoint path specified, will look for best model for seed {SEED_TO_RESUME}")
            CHECKPOINT_PATH = os.path.join(Config.ensemble_dir, f'model_seed_{SEED_TO_RESUME}_best.pth')
            if not os.path.exists(CHECKPOINT_PATH):
                CHECKPOINT_PATH = os.path.join(Config.ensemble_dir, f'last_checkpoint_seed_{SEED_TO_RESUME}.pth')

        print(f"\n📌 Resuming seed {SEED_TO_RESUME} from: {os.path.basename(CHECKPOINT_PATH) if CHECKPOINT_PATH else 'best model'}")

        # Обучаем одну модель
        model, best_f1, best_threshold = train_single_model(
            seed=SEED_TO_RESUME,
            resume_from=CHECKPOINT_PATH
        )

        # Создаём сабмит для этой модели (опционально)
        create_submit = input("\n📊 Create submission for this model? (y/n): ").lower() == 'y'
        if create_submit:
            submission = ensemble_predict_with_tta([{
                'seed': SEED_TO_RESUME,
                'f1': best_f1,
                'threshold': best_threshold,
                'path': os.path.join(Config.ensemble_dir, f'model_seed_{SEED_TO_RESUME}_best.pth')
            }])

    # ------------------------------------------------------------------------
    # НАСТРОЙКИ ДЛЯ РЕЖИМА 'train_ensemble'
    # ------------------------------------------------------------------------
    elif MODE == 'train_ensemble':
        # Настройка возобновления для каждого seed (None = начать с нуля)
        RESUME_CONFIG = {
            42: None,   # Например: '/content/drive/MyDrive/deepfake_ensemble/checkpoint_seed_42_epoch_30_f1_0.7743.pth',
            123: None,  # Продолжить с чекпоинта или None
            456: None,  # Продолжить с чекпоинта или None,
        }

        print("\n📌 Resume configuration:")
        for seed, path in RESUME_CONFIG.items():
            if path:
                print(f"   Seed {seed}: will resume from {os.path.basename(path)}")
            else:
                print(f"   Seed {seed}: will train from scratch")

        # Обучаем ансамбль
        models_info = train_ensemble(resume_config=RESUME_CONFIG)

        # Создаём сабмит с TTA
        submission = ensemble_predict_with_tta(models_info)

    # ------------------------------------------------------------------------
    # НАСТРОЙКИ ДЛЯ РЕЖИМА 'predict_only'
    # ------------------------------------------------------------------------
    elif MODE == 'predict_only':
        # Загружаем сохранённый ансамбль
        ensemble_info_path = os.path.join(Config.ensemble_dir, 'ensemble_info.pth')

        if os.path.exists(ensemble_info_path):
            ensemble_info = torch.load(ensemble_info_path, map_location='cpu')
            print(f"\n📂 Loaded ensemble info from {ensemble_info['created_at']}")
            for model_info in ensemble_info['models']:
                print(f"   Seed {model_info['seed']}: F1={model_info['f1']:.4f}")

            submission = ensemble_predict_with_tta(ensemble_info['models'])
        else:
            print("❌ Ensemble info not found! Train ensemble first.")
```
